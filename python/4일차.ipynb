# 4일차 코딩테스트 학습: 스택/큐 (Queue)

## [문제 1]: [프로그래머스 Lv.2 - 프로세스](https://school.programmers.co.kr/learn/courses/30/lessons/42587)

### 문제 설명
운영체제가 프로세스를 관리할 때, 다음 규칙에 따라 프로세스를 실행시킵니다:

1. 실행 대기 큐에서 대기 중인 프로세스 하나를 꺼냅니다.
2. 큐에 대기 중인 프로세스 중 **우선순위가 더 높은 프로세스가 하나라도 있다면**, 방금 꺼낸 프로세스를 다시 **큐의 맨 뒤에 넣습니다.**
3. 그렇지 않다면 해당 프로세스를 실행시킵니다. (한번 실행된 프로세스는 완전히 종료)

**구해야 하는 값**: 특정 위치(`location`)에 있는 프로세스가 **몇 번째로 실행되는지** return 하세요.

---

### 입출력 예시
| priorities | location | return |
| :--- | :--- | :--- |
| `[2, 1, 3, 2]` | `2` | `1` |
| `[1, 1, 9, 1, 1, 1]` | `0` | `5` |

### 입출력 예 설명
- **예시 1**:
  - 초기 대기 큐 상태: `[(0, 2), (1, 1), (2, 3), (3, 2)]` (형식: 원래 인덱스, 우선순위)
  - 첫 번째 `(0, 2)`를 꺼냈을 때, 뒤에 우선순위가 더 높은 `(2, 3)`이 있으므로 큐의 맨 뒤로 보냅니다. ➔ 대기 큐: `[(1, 1), (2, 3), (3, 2), (0, 2)]`
  - 두 번째 `(1, 1)`을 꺼냈을 때, 뒤에 우선순위가 더 높은 `(2, 3)`이 있으므로 맨 뒤로 보냅니다. ➔ 대기 큐: `[(2, 3), (3, 2), (0, 2), (1, 1)]`
  - 세 번째 `(2, 3)`을 꺼냈을 때, 뒤에 나보다 우선순위가 높은 프로세스가 없으므로 실행합니다. (1번째 실행)
  - 우리가 알고 싶었던 location `2`번 프로세스가 바로 이 `(2, 3)`이므로, 1번째 실행으로 `1`을 반환합니다.

```python
from collections import deque

def solution(priorities, location):
    queue = deque([(i, p) for i, p in enumerate(priorities)])
    order = 0  # 실행 순서
    
    while queue:
        cur = queue.popleft()  # 큐의 맨 앞 프로세스 꺼내기
        
        # TODO: 큐에 남아있는 프로세스 중 cur보다 우선순위가 높은 게 하나라도 있는지 확인하는 조건문 작성 (any 활용)
        if any(q[1] > cur[1] for q in queue):
            queue.append(cur)  # 우선순위가 높은 게 있으면 다시 큐 맨 뒤에 넣기
        else:
            order += 1  # 더 높은 게 없다면 프로세스 실행!
            
            # TODO: 방금 실행된 프로세스가 내가 찾던 location 프로세스라면 실행 순서 반환
            if cur[0] == location:
                return order

# --- 테스트 케이스 실행 ---
print("테스트 1:", solution([2, 1, 3, 2], 2))        # 기댓값: 1
print("테스트 2:", solution([1, 1, 9, 1, 1, 1], 0))  # 기댓값: 5
```

## [문제 2]: [프로그래머스 Lv.2 - 주식가격](https://school.programmers.co.kr/learn/courses/30/lessons/42584)

### 문제 설명
초 단위로 기록된 주식가격이 담긴 배열 `prices`가 주어질 때, 각 초마다 가격이 떨어지지 않은 기간은 몇 초인지를 구하여 return 하도록 solution 함수를 완성하세요.

---

### 입출력 예시
| prices | return |
| :--- | :--- |
| `[1, 2, 3, 2, 3]` | `[4, 3, 1, 1, 0]` |

### 입출력 예 설명
- 1초 시점의 가격 `1`은 5초까지 총 4초 동안 가격이 떨어지지 않고 유지됩니다.
- 2초 시점의 가격 `2`는 5초까지 총 3초 동안 가격이 떨어지지 않고 유지됩니다.
- 3초 시점의 가격 `3`은 다음 4초 시점에 `2`로 가격이 떨어집니다. 따라서 3초에서 4초가 되는 **1초 동안** 가격이 떨어지지 않은 것으로 간주하고 루프를 탈출합니다.
- 4초 시점의 가격 `2`는 5초까지 총 1초 동안 유지됩니다.
- 5초 시점의 가격 `3`은 뒤에 더이상 남은 시간이 없으므로 0초 동안 유지됩니다.

```python
from collections import deque

def solution_stock(prices):
    answer = []
    queue = deque(prices)
    
    while queue:
        price = queue.popleft() # 현재 초의 가격 꺼내기
        time = 0               # 가격이 유지된 시간
        
        # TODO: 큐에 남아있는 뒤쪽 가격들(p)을 순회하며 price와 비교하는 for문 작성
        for p in queue:
            time += 1 # 1초 지남
            
            # TODO: 만약 뒤쪽 가격 p가 현재 price보다 낮아졌다면(떨어졌다면)?
            if p < price:
                break # 비교 중단
                
        answer.append(time)
        
    return answer

# --- 테스트 케이스 실행 ---
print("테스트 결과:", solution_stock([1, 2, 3, 2, 3])) # 기댓값: [4, 3, 1, 1, 0]
```

## [문제 3]: [프로그래머스 Lv.2 - 올바른 괄호](https://school.programmers.co.kr/learn/courses/30/lessons/12909)

### 문제 설명
괄호가 바르게 짝지어졌다는 것은 `(` 문자로 열렸으면 반드시 `)` 문자로 닫혀야 한다는 뜻입니다. 예를 들어
- `()()` 또는 `(())()` 는 올바른 괄호입니다.
- `)()(` 또는 `(()(` 는 올바르지 않은 괄호입니다.

`(` 또는 `)` 로만 이루어진 문자열 `s`가 주어졌을 때, 올바른 괄호이면 `True`를, 올바르지 않은 괄호이면 `False`를 return 하는 solution 함수를 완성해 주세요.

---

### 입출력 예시
| s | answer |
| :--- | :--- |
| `"()()"` | `true` |
| `"(())()"` | `true` |
| `")()("` | `false` |
| `"(()("` | `false` |

### 입출력 예 설명
- **예시 1 (`"()()"`)**: 열리고(`(`) 닫히는(`)`) 과정이 정상적으로 두 번 반복되어 최종적으로 완전히 짝이 맞으므로 `True`를 반환합니다.
- **예시 3 (`")()("`)**: 시작부터 닫는 괄호(`)`)로 시작했기 때문에 열린 괄호가 없어 올바르지 않으므로 바로 `False`를 반환합니다.
- **예시 4 (`"(()("`)**: 여는 괄호는 총 3개인데 닫는 괄호는 1개뿐이라서 짝이 맞지 않아 최종적으로 `False`를 반환합니다.

---

### 힌트 & 생각 유도 가이드
1. **스택(Stack) 개념 활용하기**: 일반 리스트를 스택으로 사용합니다.
2. **여는 괄호 처리**: 문자열 `s`에서 문자를 하나씩 확인할 때, 여는 괄호 `(`를 만나면 스택에 넣습니다(`stack.append('(')`).
3. **닫는 괄호 처리**: 닫는 괄호 `)`를 만나면, 스택에서 여는 괄호 하나를 꺼내서 짝을 지어 없애줍니다(`stack.pop()`).
   - **주의할 점**: 만약 닫는 괄호가 나왔는데 스택이 이미 비어있다면, 짝을 맞출 여는 괄호가 없는 것이므로 즉시 올바르지 않은 괄호(`False`)로 판단하고 종료합니다.
4. **최종 확인**: 문자열 검사가 모두 끝났을 때 스택이 완전히 비어있어야 올바른 괄호입니다. 여는 괄호가 스택에 아직 남아있다면 짝이 맞지 않은 것이므로 `False`가 됩니다.

```python
def solution_parentheses(s):
    stack = []
    
    for char in s:
        if char == '(':
            # TODO: 여는 괄호는 스택에 추가합니다.
            stack.append(char)
        elif char == ')':
            # TODO: 닫는 괄호를 만났을 때, 스택이 비어있는지 확인하고 비어있으면 False를 반환합니다.
            if not stack:
                return False
            # TODO: 비어있지 않다면 스택에서 여는 괄호 하나를 꺼냅니다(pop).
            else:
                stack.pop()
            
    # TODO: 최종적으로 스택이 완전히 비어있는지 여부를 판단하여 결과를 반환합니다.
    return len(stack) == 0

# --- 테스트 케이스 실행 ---
print("테스트 1:", solution_parentheses("()()"))   # 기댓값: True
print("테스트 2:", solution_parentheses(")()("))   # 기댓값: False
```

## [문제 4]: [프로그래머스 Lv.2 - 다리를 지나는 트럭](https://school.programmers.co.kr/learn/courses/30/lessons/42583)

### 문제 설명
트럭 여러 대가 강을 가로지르는 일차선 다리를 정해진 순서대로 건너려 합니다. 모든 트럭이 다리를 건너려면 최소 몇 초가 걸리는지 알아내야 합니다. 다리에는 트럭이 최대 `bridge_length`대 올라갈 수 있으며, 다리는 `weight` 이하의 무게를 견딜 수 있습니다. 단, 다리에 완전히 오르지 않은 트럭의 무게는 무시합니다.

---

### 입출력 예시
| bridge_length | weight | truck_weights | return |
| :--- | :--- | :--- | :--- |
| `2` | `10` | `[7, 4, 5, 6]` | `8` |

### 입출력 예 설명
- 0초: 다리 상태 `[0, 0]` (길이가 2인 다리를 0으로 초기화), 대기 트럭 `[7, 4, 5, 6]`
- 1초: 7번 트럭 진입. 다리 상태 `[0, 7]`, 대기 트럭 `[4, 5, 6]`, 경과 시간 = 1초
- 2초: 7번 트럭이 1칸 전진. 다리 상태 `[7, 0]`. 다음 4번 트럭은 다리 무게 초과(7+4=11 > 10)로 못 가고 0 진입, 경과 시간 = 2초
- 3초: 7번 트럭 나감. 4번 트럭 진입. 다리 상태 `[0, 4]`, 대기 트럭 `[5, 6]`, 경과 시간 = 3초
- 4초: 4번 트럭 전진. 5번 트럭 진입 (4+5=9 <= 10). 다리 상태 `[4, 5]`, 대기 트럭 `[6]`, 경과 시간 = 4초
- 5초: 4번 트럭 나감. 6번 트럭은 다리 무게 초과(5+6=11 > 10)로 못 가고 0 진입. 다리 상태 `[5, 0]`, 경과 시간 = 5초
- 6초: 5번 트럭 나감. 6번 트럭 진입. 다리 상태 `[0, 6]`, 대기 트럭 `[]`, 경과 시간 = 6초
- 7초: 6번 트럭 전진. 다리 상태 `[6, 0]`, 경과 시간 = 7초
- 8초: 6번 트럭 완전히 나감. 다리 상태 `[0, 0]`, 모든 트럭이 다리를 건넜으므로 종료! 경과 시간 = **8초**

---

### 힌트 & 생각 유도 가이드
1. **다리를 큐(deque)로 표현하기**: 다리 길이만큼 `0`으로 채워진 큐를 만듭니다.
   - 예: `bridge = deque([0] * bridge_length)`
2. **매 초마다 다리 상황 갱신**: 
   - 1초가 흐를 때마다 다리 큐에서 맨 앞 요소 하나를 꺼냅니다 (`out_truck = bridge.popleft()`).
   - 꺼낸 트럭 무게만큼 현재 다리 위 총무게(`current_weight`)에서 빼줍니다.
3. **새로운 트럭 진입 판단**:
   - 대기 중인 트럭이 남아있고, 다음 대기 트럭이 들어왔을 때 다리 위 총무게가 한도를 넘지 않는다면:
     - 대기 트럭 큐에서 트럭을 꺼내 다리 큐의 맨 뒤에 넣습니다 (`bridge.append(next_truck)`).
     - 다리 위 총무게를 더해줍니다.
   - 무게를 초과한다면:
     - 다리 큐의 맨 뒤에 `0`을 넣습니다 (`bridge.append(0)`).

```python
from collections import deque

def solution_bridge(bridge_length, weight, truck_weights):
    time = 0
    # 1. 대기 트럭 리스트와 다리를 각각 deque로 정의
    trucks = deque(truck_weights)
    bridge = deque([0] * bridge_length)
    
    current_weight = 0  # 현재 다리 위 트럭들의 총무게
    
    # 대기 트럭이 남아있거나, 다리 위에 트럭이 아직 다 지나가지 않은 동안 반복
    while trucks or current_weight > 0:
        time += 1
        
        # 2. 다리에서 가장 앞에 있는 요소 꺼내기 (다리를 지나간 트럭)
        out_truck = bridge.popleft()
        current_weight -= out_truck
        
        if trucks:
            # TODO: 대기 트럭(trucks[0])이 들어갔을 때 무게 제한을 초과하지 않는지 확인
            if current_weight + trucks[0] <= weight:
                # TODO: 대기 큐에서 트럭을 꺼내 다리 위로 진입시키고 총무게에 더해줍니다.
                next_truck = trucks.popleft()
                bridge.append(next_truck)
                current_weight += next_truck
            else:
                # TODO: 무게 초과 시 다리에 트럭이 들어가지 못하므로 0을 다리 큐에 추가합니다.
                bridge.append(0)
                
    return time

# --- 테스트 케이스 실행 ---
print("테스트 결과:", solution_bridge(2, 10, [7, 4, 5, 6]))  # 기댓값: 8
```

## [문제 5]: [프로그래머스 Lv.2 - 짝지어 제거하기](https://school.programmers.co.kr/learn/courses/30/lessons/12973)

### 문제 설명
짝지어 제거하기는, 알파벳 소문자로 이루어진 문자열을 가지고 시작합니다. 먼저 문자열에서 같은 알파벳이 2개 붙어 있는 짝을 찾습니다. 그다음, 그 둘을 제거한 뒤, 앞뒤로 문자열을 이어 붙입니다. 이 과정을 반복해서 문자열을 모두 제거할 수 있다면 1을, 아닐 경우 0을 return 하는 함수를 완성해 주세요.

예를 들어, 문자열 `s = baabaa` 라면
b **aa** baa ➔ **bb** aa ➔ **aa** ➔ empty
순서대로 모두 제거할 수 있으므로 1을 반환합니다.

---

### 입출력 예시
| s | result |
| :--- | :--- |
| `"baabaa"` | `1` |
| `"cdcd"` | `0` |

### 입출력 예 설명
- **예시 1 (`"baabaa"`)**:
  1. 1번째 문자 'b' 확인: 스택이 비어있으므로 추가합니다. ➔ 스택: `['b']`
  2. 2번째 문자 'a' 확인: 스택의 맨 위('b')와 현재 문자('a')가 다르므로 추가합니다. ➔ 스택: `['b', 'a']`
  3. 3번째 문자 'a' 확인: 스택의 맨 위('a')와 현재 문자('a')가 같습니다! 짝이 맞으므로 스택에서 기존 'a'를 제거합니다. ➔ 스택: `['b']`
  4. 4번째 문자 'b' 확인: 스택의 맨 위('b')와 현재 문자('b')가 같습니다! 짝이 맞으므로 스택에서 기존 'b'를 제거합니다. ➔ 스택: `[]`
  5. 5번째 문자 'a' 확인: 스택이 비어있으므로 추가합니다. ➔ 스택: `['a']`
  6. 6번째 문자 'a' 확인: 스택의 맨 위('a')와 현재 문자('a')가 같습니다! 짝이 맞으므로 스택에서 기존 'a'를 제거합니다. ➔ 스택: `[]`
  - 모든 문자를 확인한 후 스택이 완전히 비어있으므로 전부 제거 가능 ➔ return `1`

- **예시 2 (`"cdcd"`)**:
  1. 'c' 진입 ➔ 스택: `['c']`
  2. 'd' 진입 (c와 다름) ➔ 스택: `['c', 'd']`
  3. 'c' 진입 (d와 다름) ➔ 스택: `['c', 'd', 'c']`
  4. 'd' 진입 (c와 다름) ➔ 스택: `['c', 'd', 'c', 'd']`
  - 같은 알파벳이 이웃하여 짝을 짓는 경우가 한 번도 없으므로 전혀 지워지지 않아 스택에 남습니다. ➔ return `0`

### 힌트 & 생각 유도 가이드
1. **스택의 활용성**: 이웃한 같은 문자끼리 짝을 짓는 것은 직전에 들어온 원소와 현재 원소를 비교하는 **스택(Stack)**의 특징을 가집니다.
2. **흐름 설계**:
   - 문자열 `s`의 문자들을 순회합니다.
   - 스택이 **비어있지 않고**, 스택의 **맨 위 원소(마지막 추가된 값)**가 **현재 문자**와 같다면:
     - 짝이 완성되었으므로 스택의 맨 위 원소를 제거합니다 (`stack.pop()`).
   - 그렇지 않다면(스택이 비어있거나, 맨 위 원소와 현재 문자가 다르다면):
     - 현재 문자를 스택에 넣습니다.
3. **결과 도출**: 모든 순회가 끝났을 때 스택이 완전히 비어있으면 `1`을, 비어있지 않고 원소가 남아있으면 `0`을 반환합니다.

```python
def solution_remove_pairs(s):
    stack = []
    
    for char in s:
        # 스택이 비어있지 않고, 마지막 글자가 현재 글자와 같은지 확인
        if stack and stack[-1] == char:
            stack.pop()
        else:
            stack.append(char)
        
    # 스택이 비어있는지 판단하여 1 또는 0을 반환
    if len(stack) == 0:
        return 1
    else:
        return 0

# --- 테스트 케이스 실행 ---
print("테스트 1:", solution_remove_pairs("baabaa"))  # 기댓값: 1
print("테스트 2:", solution_remove_pairs("cdcd"))    # 기댓값: 0
```

## [문제 6]: [프로그래머스 Lv.2 - 괄호 회전하기](https://school.programmers.co.kr/learn/courses/30/lessons/76502)

대괄호 `[]`, 중괄호 `{}`, 소괄호 `()`로 이루어진 문자열 `s`가 주어집니다. 이 문자열을 왼쪽으로 $x$칸 만큼 회전시켰을 때 ($0 \le x < \text{len}(s)$), 올바른 괄호 문자열이 되도록 하는 $x$의 개수를 구하는 함수를 완성해 주세요.

### 입출력 예 설명
- **예시 1 (`"[](){}"`)**:
  - $x = 0$ (0칸 회전): `"[](){}"` ➔ 대괄호, 소괄호, 중괄호가 순서대로 모두 올바르게 열리고 닫혀 짝이 맞으므로 올바른 괄호입니다. (올바름 +1)
  - $x = 1$ (1칸 회전): `"](){}["` ➔ 닫는 괄호 `]`로 시작했기 때문에 무조건 올바르지 않은 괄호입니다. (무시)

- $x = 3$ (3칸 회전): `"){}[ ]("` ➔ 닫는 괄호 `)`로 시작했기 때문에 올바르지 않습니다. (무시)
  - $x = 4$ (4칸 회전): `"{}[ ]()"` ➔ 중괄호, 대괄호, 소괄호의 짝이 모두 올바르게 맞습니다. (올바름 +1)
  - $x = 5$ (5칸 회전): `"}[ ](){`"` ➔ 닫는 괄호 `}`로 시작했기 때문에 올바르지 않습니다. (무시)
  - 따라서 올바른 괄호 문자열이 되는 회전 횟수 $x$는 0, 2, 4로 총 **3개**이므로 `3`을 반환합니다.

### 힌트 & 생각 유도 가이드
1. **문자열 회전 방법**:
   - 문자열 슬라이싱 `s[x:] + s[:x]`를 사용하면 $x$만큼 왼쪽으로 회전한 문자열을 간단하게 만들 수 있습니다.
   - 혹은 큐(`deque`)에 글자들을 다 담아두고 `rotate(-1)`을 매번 호출하여 한 칸씩 왼쪽으로 넘길 수도 있습니다.
2. **여러 종류의 괄호 짝 짓기 (스택)**:
   - 괄호의 종류가 `()`, `[]`, `{}` 세 종류이므로, 딕셔너리(`mapping`)를 활용해 매칭 정보를 저장해두면 편리합니다.
   - 예: `mapping = {')': '(', ']': '[', '}': '{'}`
   - 닫는 괄호가 나왔을 때 스택에서 꺼낸 값(`stack.pop()`)이 딕셔너리에 매핑된 올바른 짝의 여는 괄호와 같은지 비교합니다.
3. **흐름 설계**:
   - 전체 길이 `len(s)`만큼 루프를 돌며 매 회전된 문자열을 구합니다.
   - 각 회전된 문자열이 올바른 괄호인지 판별하는 함수(또는 코드)를 실행합니다.
   - 올바른 괄호로 확인될 때마다 카운트(`count += 1`)를 세고 최종 카운트를 반환합니다.

```python
def check_valid(string):
    # TODO: 괄호 매핑 딕셔너리 정의
    mapping = {')': '(', ']': '[', '}': '{'}
    stack = []
    
    # TODO: string의 각 글자를 순회하며 올바른 괄호 구조인지 체크하는 로직을 작성하세요.
    # 힌트: 여는 괄호('(', '[', '{')는 스택에 넣고,
    # 닫는 괄호(')', ']', '}')가 나오면 스택이 비어있는지 확인하고,
    # 비어있지 않다면 스택에서 꺼낸 값이 해당 닫는 괄호의 올바른 짝(mapping[char])과 같은지 확인하세요.
    for char in string:
        if char in mapping.values():
            stack.append(char)
        else:
            if not stack:
                return False
            if mapping[char] != stack.pop():
                return False
    return len(stack) == 0

def solution_rotate(s):
    answer = 0
    n = len(s)
    
    # TODO: 0부터 n-1까지 회전수 x를 늘리며 문자열을 회전시키고 check_valid를 호출하세요.
    # 회전된 문자열 = s[x:] + s[:x]
    for x in range(n):
        rotate_s = s[x:] +s[:x]
        if check_valid(rotate_s):
            answer +=1        
    return answer

# --- 테스트 케이스 실행 ---
print("테스트 결과 1:", solution_rotate("[](){}")) # 기댓값: 3
print("테스트 결과 2:", solution_rotate("}]()[{")) # 기댓값: 2
```
