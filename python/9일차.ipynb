## 문제 1: 탐욕법(Greedy) - 구명보트 (Level 2)

- **문제 출처**: [프로그래머스 구명보트](https://school.programmers.co.kr/learn/courses/30/lessons/42885)

---

### 문제 설명

무인도에 갇힌 사람들을 구명보트를 이용하여 구출하려 합니다. 구명보트는 작은 크기여서 최대 2명씩 밖에 탈 수 없고, 무게 제한도 있습니다.

사람들의 몸무게를 담은 배열 `people`과 구명보트의 무게 제한 `limit`가 매개변수로 주어질 때, 모든 사람을 구출하기 위해 필요한 구명보트 개수의 최솟값을 return 하도록 solution 함수를 작성해주세요.

---

### 입출력 예시 및 작동 흐름 설명

| people | limit | return |
| --- | --- | --- |
| [70, 50, 80, 50] | 100 | 3 |
| [70, 80, 50] | 100 | 3 |

#### 진행 과정 및 상세 작동 흐름 (people = [70, 50, 80, 50], limit = 100 예시)
1. 정렬된 몸무게: [50, 50, 70, 80]
2. 가장 가벼운 사람(50kg)과 가장 무거운 사람(80kg)의 합 = 130kg > 100kg (같이 못 탐 -> 80kg 혼자 탑승, 보트 1개)
3. 가장 가벼운 사람(50kg)과 그다음 무거운 사람(70kg)의 합 = 120kg > 100kg (같이 못 탐 -> 70kg 혼자 탑승, 보트 2개)
4. 남은 50kg 두 사람의 합 = 100kg <= 100kg (같이 탑승, 보트 3개)
5. 따라서 최소 3개의 구명보트가 필요합니다.


```python
def solution(people, limit):
    people.sort()
    start = 0
    end = len(people)-1
    answer = 0

    while start <= end:
        if people[start] + people[end] <= limit:
            start +=1
        end -=1
        answer +=1
    return answer
```

## 문제 2: 정렬(Sorting) - K번째수 (Level 1)

- **문제 출처**: [프로그래머스 K번째수](https://school.programmers.co.kr/learn/courses/30/lessons/42748)

---

### 문제 설명

배열 `array`의 i번째 숫자부터 j번째 숫자까지 자르고 정렬했을 때, k번째에 있는 수를 구하려 합니다.

예를 들어 `array`가 [1, 5, 2, 6, 3, 7, 4], i = 2, j = 5, k = 3이라면
1. `array`의 2번째부터 5번째까지 자르면 [5, 2, 6, 3]입니다.
2. 1에서 나온 배열을 정렬하면 [2, 3, 5, 6]입니다.
3. 2에서 나온 배열의 3번째 숫자는 5입니다.

배열 `array`, [i, j, k]를 원소로 가진 2차원 배열 `commands`가 매개변수로 주어질 때, `commands`의 모든 원소에 대해 앞서 설명한 연산을 적용했을 때의 결과를 배열에 담아 return 하도록 solution 함수를 작성해주세요.

---

### 입출력 예시 및 작동 흐름 설명

| array | commands | return |
| --- | --- | --- |
| [1, 5, 2, 6, 3, 7, 4] | [[2, 5, 3], [4, 4, 1], [1, 7, 3]] | [5, 6, 3] |

#### 진행 과정 및 상세 작동 흐름
1. [2, 5, 3] 연산: 2번째~5번째인 [5, 2, 6, 3] 자르고 정렬 -> [2, 3, 5, 6] 중 3번째 값 = 5
2. [4, 4, 1] 연산: 4번째~4번째인 [6] 자르고 정렬 -> [6] 중 1번째 값 = 6
3. [1, 7, 3] 연산: 1번째~7번째인 전체 자르고 정렬 -> [1, 2, 3, 4, 5, 6, 7] 중 3번째 값 = 3
4. 결과 [5, 6, 3] 반환


```python
def solution(array, commands):
    answer = []

    for i,j,k in commands:
        a = array[i-1:j]
        a.sort()
        b = a[k-1]
        answer.append(b)    
    return answer
```

## 문제 3: 정렬(Sorting) - 가장 큰 수 (Level 2)

- **문제 출처**: [프로그래머스 가장 큰 수](https://school.programmers.co.kr/learn/courses/30/lessons/42746)

---

### 문제 설명

0 또는 양의 정수가 주어졌을 때, 정수를 이어 붙여 만들 수 있는 가장 큰 수를 알고 싶습니다.

예를 들어, 주어진 정수가 [6, 10, 2]라면 [6102, 6210, 1062, 1026, 2610, 2106]를 만들 수 있고, 이 중 가장 큰 수는 6210입니다.

0 또는 양의 정수가 담긴 배열 `numbers`가 매개변수로 주어질 때, 순서를 바꾸어 만들 수 있는 가장 큰 수를 문자열로 바꾼 정답을 return 하도록 solution 함수를 작성해주세요.

---

### 입출력 예시 및 작동 흐름 설명

| numbers | return |
| --- | --- |
| [6, 10, 2] | "6210" |
| [3, 30, 34, 5, 9] | "9534330" |

#### 진행 과정 및 상세 작동 흐름 (numbers = [3, 30, 34, 5, 9] 예시)
1. 각 숫자를 문자로 이어붙였을 때 더 큰 조합이 오도록 정렬 규칙을 세웁니다.
2. 9, 5, 34, 3, 30 순으로 배치되었을 때 문자열 "9534330"이 가장 큰 수가 됩니다.


```python
def solution(numbers):
    answer = ""
    result = []
    for i in numbers:
        result.append(str(i))
    result.sort(key=lambda x: x*3, reverse=True)
    answer = "".join(result)
    if answer[0] == "0":
        return "0"
    return answer
```
