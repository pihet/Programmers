```python
"""
[프로그래머스 Lv.2 - 기능개발 (스택/큐)]

- 문제 설명:
  프로그래머스 팀에서는 기능 개선 작업을 수행 중입니다. 
  각 기능은 진도가 100%일 때 서비스에 반영할 수 있습니다.
  각 기능의 현재 개발 진도(progresses)와 개발 속도(speeds)가 주어집니다.
  
  뒤에 있는 기능이 앞에 있는 기능보다 먼저 개발될 수 있지만, 
  앞에 있는 기능이 배포될 때 함께 배포되어야 합니다.
  각 배포마다 몇 개의 기능이 배포되는지 return 하세요.

- 입출력 예시:
  progresses = [93, 30, 55]
  speeds     = [1, 30, 5]
  
  1) 1일째: 94, 60, 60
  2) 7일째: 100, 100, 95 -> 첫 번째(100)와 두 번째(100) 기능 완성! -> 2개 배포!
  3) 9일째: 95% -> 100%  -> 세 번째(100) 기능 완성! -> 1개 배포!
  
  -> return [2, 1]

- 힌트/단계별 생각하기:
  1. 각 작업이 완성되기까지 '며칠'이 걸리는지 먼저 계산해 봅니다.
     - (100 - progress) / speed
     - 예: 93%에 속도 1이면 (100-93)/1 = 7일 소요
     - 예: 30%에 속도 30이면 (100-30)/30 = 2.33..일 -> 3일 소요
  2. 완성까지 걸리는 일수를 큐(deque)에 넣습니다. -> deque([7, 3, 9])
  3. 큐에서 맨 앞의 일수(7일)를 꺼내고, 그 7일보다 일찍 끝나거나 같은 일수들(3일)을 큐에서 함께 꺼내어 개수를 셉니다!
"""

from collections import deque
import math

def solution(progresses, speeds):
    answer = []
    
    # 1. 각 작업의 완료 소요 일수를 구해 deque에 넣기
    # math.ceil((100 - p) / s) 사용 가능
    days = deque([math.ceil((100 - p) / s) for p, s in zip(progresses, speeds)])
    
    # 2. 큐가 빌 때까지 반복하며 배포 개수 세기
    while days:
        first = days.popleft() # 가장 앞에 있는 작업 소요 일수
        count = 1
        while days and days[0] <= first:
            days.popleft()
            count +=1 
        answer.append(count)
        
    return answer
```
