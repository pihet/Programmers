## 문제 1: 해시 - 베스트앨범 (Level 3)

- **문제 출처**: [프로그래머스 베스트앨범](https://school.programmers.co.kr/learn/courses/30/lessons/42579)

---

### 문제 설명

스트리밍 사이트에서 장르 별로 가장 많이 재생된 노래를 두 개씩 모아 베스트 앨범을 출시하려 합니다. 노래는 고유 번호로 구분하며, 노래를 수록하는 기준은 다음과 같습니다.

1. 속한 노래가 많이 재생된 장르를 먼저 수록합니다.
2. 장르 내에서 많이 재생된 노래를 먼저 수록합니다.
3. 장르 내에서 재생 횟수가 같은 노래 중에서는 고유 번호가 낮은 노래를 먼저 수록합니다.

노래의 장르를 나타내는 문자열 배열 `genres`와 노래별 재생 횟수를 나타내는 정수 배열 `plays`가 주어질 때, 베스트 앨범에 들어갈 노래의 고유 번호를 순서대로 return 하도록 solution 함수를 완성하세요.

---

### 입출력 예시 및 작동 흐름 설명

| genres | plays | return |
| --- | --- | --- |
| ["classic", "pop", "classic", "classic", "pop"] | [500, 600, 150, 800, 2500] | [4, 1, 3, 0] |

#### 진행 과정 및 상세 작동 흐름
1. classic 장르는 총 1,450회 재생되었고, pop 장르는 총 3,100회 재생되었으므로 pop 장르의 노래를 먼저 수록합니다.
2. pop 장르 노래 중 4번 고유번호(2,500회)가 1번(600회)보다 많이 재생되어 [4, 1] 순으로 수록합니다.
3. classic 장르 노래 중 3번 고유번호(800회)가 0번(500회)보다 많이 재생되어 [3, 0] 순으로 수록합니다.
4. 최종적으로 [4, 1, 3, 0]이 베스트 앨범의 트랙으로 구성됩니다.

```python
def solution(genres, plays):
    answer = []
    ge = {}
    total_plays = {}
    for i,(g,p) in enumerate(zip(genres,plays)):
        if g not in ge:
            ge[g] = [(p,i)]
            total_plays[g] = total_plays.get(g, 0) + p
        else:
            ge[g].append((p,i))
            total_plays[g] = total_plays.get(g, 0) + p
            
    sort_re = sorted(total_plays, key = lambda x : total_plays[x], reverse=True)
    for g in sort_re:
        sorted_song = sorted(ge[g], key=lambda x: (-x[0], x[1]))
        for song in sorted_song[:2]:
            answer.append(song[1])
    return answer
        

```

## 문제 2: 깊이/너비 우선 탐색(DFS/BFS) - 단어 변환 (Level 3)

- **문제 출처**: [프로그래머스 단어 변환](https://school.programmers.co.kr/learn/courses/30/lessons/43163)

---

### 문제 설명

두 개의 단어 `begin`, `target`과 단어의 집합 `words`가 있습니다. 아래와 같은 규칙을 이용하여 `begin`에서 `target`으로 변환하는 가장 짧은 변환 과정을 찾으려고 합니다.

1. 한 번에 한 개의 알파벳만 바꿀 수 있습니다.
2. `words`에 있는 단어로만 변환할 수 있습니다.

예를 들어 `begin`이 "hit", `target`이 "cog", `words`가 ["hot", "dot", "dog", "lot", "log", "cog"]라면 "hit" -> "hot" -> "dot" -> "dog" -> "cog"와 같이 4단계만에 변환할 수 있습니다.

최소 몇 단계의 과정을 거쳐 `begin`을 `target`으로 변환할 수 있는지 return 하도록 solution 함수를 작성해주세요. 변환할 수 없는 경우에는 0을 return 합니다.

---

### 입출력 예시 및 작동 흐름 설명

| begin | target | words | return |
| --- | --- | --- | --- |
| "hit" | "cog" | ["hot", "dot", "dog", "lot", "log", "cog"] | 4 |
| "hit" | "cog" | ["hot", "dot", "dog", "lot", "log"] | 0 |

#### 진행 과정 및 상세 작동 흐름 (첫 번째 예시)
1. "hit"에서 글자 1개만 다른 "hot"으로 변환합니다. (1단계)
2. "hot"에서 글자 1개만 다른 "dot" 또는 "lot"으로 변환할 수 있습니다. "dot"으로 이동합니다. (2단계)
3. "dot"에서 "dog"로 변환합니다. (3단계)
4. "dog"에서 목표 단어인 "cog"로 변환 완료합니다. (4단계)
5. 가장 짧은 변환 과정의 단계 수인 4를 반환합니다.

```python
from collections import deque
def solution(begin, target, words):
    visited = set([begin])
    q = deque([(begin,0)])

    if target not in words:
        return 0
    
    while q:
        i,j = q.popleft()
        if i == target:
            return j

        for w in words:
            if w not in visited:
                diff = 0
                for c1,c2 in zip(i,w):
                    if c1 != c2:
                        diff +=1        
                if diff == 1:
                    visited.add(w)
                    q.append((w,j+1))
    return 0
```

## 문제 3: 완전탐색(Brute-force) - 전력망을 둘로 나누기 (Level 2)

- **문제 출처**: [프로그래머스 전력망을 둘로 나누기](https://school.programmers.co.kr/learn/courses/30/lessons/86971)

---

### 문제 설명

n개의 송전탑이 전선을 통해 트리 형태로 연결되어 있습니다. 당신은 이 전선들 중 하나를 끊어서 현재의 전력망 네트워크를 2개로 분할하려고 합니다. 이때, 두 전력망이 갖게 되는 송전탑 개수의 차이(절댓값)를 최소화하고자 합니다.

송전탑의 개수 `n`, 그리고 전선 정보 `wires`가 매개변수로 주어질 때, 전선들 중 하나를 끊어서 송전탑 개수의 차이의 최솟값을 return 하도록 solution 함수를 완성해주세요.

---

### 입출력 예시 및 작동 흐름 설명

| n | wires | return |
| --- | --- | --- |
| 9 | [[1,3],[2,3],[3,4],[4,5],[4,6],[4,7],[7,8],[7,9]] | 1 |
| 4 | [[1,2],[2,3],[3,4]] | 0 |
| 7 | [[1,2],[2,7],[3,7],[3,4],[4,5],[6,7]] | 1 |

#### 진행 과정 및 상세 작동 흐름 (첫 번째 예시)
1. [3, 4] 전선을 끊으면 하나의 전력망은 {1, 2, 3} (3개), 다른 전력망은 {4, 5, 6, 7, 8, 9} (6개)로 분할됩니다.
2. 이때 두 전력망의 송전탑 개수 차이는 |3 - 6| = 3입니다.
3. 만약 [4, 7] 전선을 끊으면 {1, 2, 3, 4, 5, 6} (6개)와 {7, 8, 9} (3개)로 차이는 3입니다.
4. 여러 전선 중 차이가 최솟값이 되는 전선은 [4, 7] 또는 [3, 4] 등을 끊을 때 나오는 최솟값 3입니다.

```python
from collections import deque
def solution(n, wires):
    min_diff = n
    for v1,v2 in wires:
        visited = set([v1])
        q = deque([v1])
        count = 1

        while q:
            curr = q.popleft()
            for a,b in wires:
                if (a == v1 and b == v2) or (a == v2 and b == v1):
                    continue
                if a == curr and b not in visited:
                    visited.add(b)
                    q.append(b)
                    count +=1
                elif b == curr and a not in visited:
                    visited.add(a)
                    q.append(a)
                    count +=1
        diff = abs(count - (n - count))
        min_diff = min(min_diff,diff)
    return min_diff
```
