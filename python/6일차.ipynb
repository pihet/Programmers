## [문제 1]: [프로그래머스 Lv.2 - 더 맵다](https://school.programmers.co.kr/learn/courses/30/lessons/42626)

### 문제 설명
매운 것을 좋아하는 Leo는 모든 음식의 스코빌 지수를 K 이상으로 만들고 싶습니다. 모든 음식의 스코빌 지수를 K 이상으로 만들기 위해 Leo는 스코빌 지수가 가장 낮은 두 개의 음식을 아래와 같이 특별한 방법으로 섞어 새로운 음식을 만듭니다.

```
섞은 음식의 스코빌 지수 = 가장 맵지 않은 음식의 스코빌 지수 + (두 번째로 맵지 않은 음식의 스코빌 지수 * 2)
```

Leo는 모든 음식의 스코빌 지수가 K 이상이 될 때까지 반복하여 섞습니다.
Leo가 가진 음식의 스코빌 지수를 담은 배열 `scoville`과 원하는 스코빌 지수 `K`가 주어질 때, 모든 음식의 스코빌 지수를 K 이상으로 만들기 위해 섞어야 하는 최소 횟수를 return 하도록 solution 함수를 작성해주세요.

#### 제한 사항
- scoville의 길이는 2 이상 1,000,000 이하입니다.
- K는 0 이상 1,000,000,000 이하입니다.
- scoville의 원소는 각각 0 이상 1,000,000 이하입니다.
- 모든 음식의 스코빌 지수를 K 이상으로 만들 수 없는 경우에는 -1을 return 합니다.

---

#### 입출력 예시
| scoville | K | return |
| :--- | :--- | :--- |
| `[1, 2, 3, 9, 10, 12]` | `7` | `2` |

#### 입출력 예 설명
- **예시 1**:
  - 초기 음식 스코빌 지수: `[1, 2, 3, 9, 10, 12]`
  - 1차 섞기: 가장 맵지 않은 1과 두 번째 2를 섞음 -> `1 + (2 * 2) = 5` 생성.
  - 남은 음식 스코빌: `[3, 5, 9, 10, 12]` (최소값 3은 여전히 7 미만)
  - 2차 섞기: 가장 맵지 않은 3과 두 번째 5를 섞음 -> `3 + (5 * 2) = 13` 생성.
  - 남은 음식 스코빌: `[9, 10, 12, 13]` (모든 음식의 스코빌 지수가 7 이상이 됨)
  - 총 2회 섞었으므로 `2`를 반환합니다.

---

#### 힌트 & 생각 유도 가이드
- **최소 힙 (Min Heap)의 필요성**:
  - 매 단계마다 가장 작은 값 2개를 꺼내서 계산한 후 다시 넣어야 합니다.
  - 일반 리스트에서 `min()`을 사용하거나 매번 정렬(`sort()`)하면 O(N^2) 또는 O(N^2 log N)의 시간 복잡도가 발생하여 효율성 테스트에서 통과하지 못합니다.
  - 파이썬의 `heapq` 모듈을 사용하면 리스트를 O(N)으로 힙 구조로 변환할 수 있고, 삽입/삭제를 O(log N) 만에 수행할 수 있습니다.
- **단계별 의사코드 (Pseudocode)**:
  1. `import heapq` 후 `heapq.heapify(scoville)`로 힙 구조 변환.
  2. 섞은 횟수 카운터 `count = 0` 초기화.
  3. `scoville[0]`(가장 작은 값)이 `K` 미만인 동안 루프 수행:
     - 만약 남아있는 음식의 개수(`len(scoville)`)가 2개 미만이라면, 더 이상 섞을 수 없으므로 `-1` 반환.
     - `heapq.heappop(scoville)`으로 첫 번째 최솟값 추출.
     - `heapq.heappop(scoville)`으로 두 번째 최솟값 추출.
     - 새로운 스코빌 = 첫 번째 + (두 번째 * 2) 계산 후 `heapq.heappush(scoville, new_scoville)` 실행.
     - `count += 1` 증가.
  4. 루프 종료 후 `count` 반환.


````python
import heapq

def solution(scoville, K):
    heapq.heapify(scoville)
    count = 0 

    while scoville[0] < K :
        if len(scoville) < 2:
            return -1
        first = heapq.heappop(scoville)
        second = heapq.heappop(scoville)
        new = first + (second * 2)
        heapq.heappush(scoville,new)
        count += 1
    return count
````

## [문제 3]: [프로그래머스 Lv.1 - K번째수](https://school.programmers.co.kr/learn/courses/30/lessons/42748)

### 문제 설명
배열 `array`의 i번째 숫자부터 j번째 숫자까지 자르고 정렬했을 때, k번째에 있는 수를 구하려 합니다.

예를 들어 `array`가 `[1, 5, 2, 6, 3, 7, 4]`, i = 2, j = 5, k = 3이라면
1. `array`의 2번째부터 5번째까지 자르면 `[5, 2, 6, 3]`입니다.
2. 1에서 나온 배열을 정렬하면 `[2, 3, 5, 6]`입니다.
3. 2에서 나온 배열의 3번째 숫자는 `5`입니다.

배열 `array`와 `[i, j, k]`를 원소로 가진 2차원 배열 `commands`가 매개변수로 주어질 때, `commands`의 모든 원소에 대해 앞서 설명한 연산을 적용했을 때의 결과를 배열에 담아 return 하도록 solution 함수를 작성해주세요.

#### 제한 사항
- array의 길이는 1 이상 100 이하입니다.
- array의 각 원소는 1 이상 100 이하입니다.
- commands의 길이는 1 이상 50 이하입니다.
- commands의 각 원소는 길이가 3인 배열로 [i, j, k] 모양입니다.

---

#### 입출력 예시
| array | commands | return |
| :--- | :--- | :--- |
| `[1, 5, 2, 6, 3, 7, 4]` | `[[2, 5, 3], [4, 4, 1], [1, 7, 3]]` | `[5, 6, 3]` |

#### 입출력 예 설명
- **예시 1**:
  - `[1, 5, 2, 6, 3, 7, 4]`를 2번째부터 5번째까지 자른 후 정렬: `[2, 3, 5, 6]` -> 3번째 값은 `5`.
  - `[1, 5, 2, 6, 3, 7, 4]`를 4번째부터 4번째까지 자른 후 정렬: `[6]` -> 1번째 값은 `6`.
  - `[1, 5, 2, 6, 3, 7, 4]`를 1번째부터 7번째까지 자른 후 정렬: `[1, 2, 3, 4, 5, 6, 7]` -> 3번째 값은 `3`.
  - 최종 결과: `[5, 6, 3]`

---

#### 힌트 & 생각 유도 가이드
- **슬라이싱(Slicing)과 인덱싱 주의점**:
  - 파이썬에서 리스트 인덱스는 0부터 시작하지만, 문제에서 `i`, `j`, `k`는 1번째부터 시작하는 사람 기준 인덱스입니다.
  - 따라서 `i`번째부터 `j`번째까지 슬라이싱하려면 `array[i-1:j]`로 슬라이싱해야 합니다.
  - 슬라이싱된 자른 리스트를 정렬(`sorted()`)한 뒤 `k-1`번째 인덱스의 값을 가져옵니다.


````python
def solution(array, commands):
    answer = []
    for i in commands:
        a = i[0]
        b = i[1]
        c = i[2]
        temp = sorted(array[a-1:b])
        result = temp[c-1]
        answer.append(result)
    return answer
````
