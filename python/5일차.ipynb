## [문제 3]: [프로그래머스 Lv.1 - 완주하지 못한 선수](https://school.programmers.co.kr/learn/courses/30/lessons/42576)

### 문제 설명
수많은 마라톤 선수들이 마라톤에 참여하였습니다. 단 한 명의 선수를 제외하고는 모든 선수가 마라톤을 완주하였습니다.

마라톤에 참여한 선수들의 이름이 담긴 배열 `participant`와 완주한 선수들의 이름이 담긴 배열 `completion`이 주어질 때, 완주하지 못한 선수의 이름을 return 하도록 solution 함수를 작성해주세요.

#### 제한 사항
- 마라톤 경기에 참여한 선수의 수는 1명 이상 100,000명 이하입니다.
- completion의 길이는 participant의 길이보다 1 작습니다.
- 참가자의 이름은 1개 이상 20개 이하의 알파벳 소문자로 이루어져 있습니다.
- 참가자 중에는 동명이인이 있을 수 있습니다.

---

#### 입출력 예시
| participant | completion | return |
| :--- | :--- | :--- |
| `["leo", "kiki", "eden"]` | `["eden", "kiki"]` | `"leo"` |
| `["marina", "josipa", "nikola", "vinko", "filipa"]` | `["josipa", "filipa", "marina", "nikola"]` | `"vinko"` |
| `["mislav", "stanko", "mislav", "ana"]` | `["stanko", "ana", "mislav"]` | `"mislav"` |

#### 입출력 예 설명
- **예시 1**:
  - 참여자 중 "leo", "kiki", "eden"이 있고 완주자 중 "eden", "kiki"가 있습니다.
  - 완주하지 못한 선수는 "leo"입니다.
- **예시 2**:
  - 참여자 중 "vinko"는 완주자 명단에 없으므로 완주하지 못했습니다.
- **예시 3**:
  - 참여자 중 "mislav"가 두 명(동명이인)이 있고, 완주자 중 한 명만 완주했습니다.
  - 따라서 완주하지 못한 한 명의 "mislav"를 반환합니다.

---

#### 힌트 & 생각 유도 가이드
- **해시(딕셔너리)를 이용한 개수 카운팅**:
  - 참여자 리스트(`participant`)를 순회하며 각 이름별 등장 횟수를 딕셔너리에 누적합니다. (예: `{"mislav": 2, "stanko": 1}`)
  - 완주자 리스트(`completion`)를 순회하며 딕셔너리에서 해당 이름의 개수를 1씩 차감합니다.
  - 마지막에 딕셔너리를 순회하면서 남은 개수가 1개 이상인 이름을 찾아 반환합니다.
- **Counter 객체 활용**:
  - 파이썬 `collections.Counter`를 사용하면 카운터 객체끼리 뺄셈 연산(`Counter(A) - Counter(B)`)이 가능합니다. 이 뺄셈 결과를 통해 완주하지 못한 한 명을 쉽게 찾을 수 있습니다.

```python
def solution(participant, completion):
    name = {}
    result = ""
    for i in participant:
        if i not in name:
            name[i] =1
        else:
            name[i] +=1

    for i in completion:
            name[i] -= 1

    for i in name:
        if name[i] == 1:
            return i
    
           
```

## [문제 4]: [프로그래머스 Lv.1 - 폰켓몬](https://school.programmers.co.kr/learn/courses/30/lessons/1845)

### 문제 설명
당신은 N마리의 폰켓몬 중에서 N/2마리를 가져갈 수 있습니다. 폰켓몬들은 종류에 따라 번호가 부여되어 있으며, 같은 종류의 폰켓몬은 같은 번호를 가지고 있습니다.
당신은 최대한 다양한 종류의 폰켓몬을 가지길 원합니다.
N마리 폰켓몬의 종류 번호가 담긴 배열 `nums`가 주어질 때, N/2마리의 폰켓몬을 선택하는 방법 중, 가장 많은 종류의 폰켓몬을 선택하는 방법을 찾아 그때의 폰켓몬 종류 번호의 개수를 return 하도록 solution 함수를 완성해주세요.

#### 제한 사항
- nums는 폰켓몬의 종류 번호가 담긴 1차원 배열입니다.
- nums의 길이(N)는 1 이상 10,000 이하의 자연수이며, 항상 짝수입니다.
- 폰켓몬의 종류 번호는 1 이상 200,000 이하의 자연수입니다.

---

#### 입출력 예시
| nums | return |
| :--- | :--- |
| `[3, 1, 2, 3]` | `2` |
| `[3, 3, 3, 2, 2, 4]` | `3` |
| `[3, 3, 3, 2, 2, 2]` | `2` |

#### 입출력 예 설명
- **예시 1**:
  - 폰켓몬이 총 4마리 있으므로 우리는 2마리를 선택할 수 있습니다.
  - 종류는 1번, 2번, 3번으로 총 3가지입니다. 이 중 2마리를 고를 때 가장 많은 종류를 선택하는 방법은 2가지 종류를 고르는 것입니다. 따라서 `2`를 반환합니다.
- **예시 2**:
  - 폰켓몬이 총 6마리 있으므로 우리는 3마리를 선택할 수 있습니다.
  - 종류는 2번, 3번, 4번으로 총 3가지입니다. 이 3가지를 모두 고르면 가장 다양한 종류를 가질 수 있습니다. 따라서 `3`를 반환합니다.
- **예시 3**:
  - 폰켓몬이 총 6마리이므로 3마리를 선택할 수 있지만, 종류는 2번, 3번으로 2가지뿐입니다.
  - 따라서 아무리 다양하게 골라도 최대 2가지 종류만 가져갈 수 있으므로 `2`를 반환합니다.

---

#### 힌트 & 생각 유도 가이드
- **핵심 아이디어: 중복 제거와 비교**:
  - 가져갈 수 있는 최대 폰켓몬 마리 수는 전체 수의 절반인 `N // 2` 마리입니다.
  - 폰켓몬 배열 `nums`에서 중복을 제거하면 총 몇 종류의 폰켓몬이 존재하는지 알 수 있습니다. 파이썬의 집합(`set`) 자료형을 이용하면 중복을 쉽게 제거할 수 있습니다. (종류 수 = `len(set(nums))`)
  - 만약 존재하는 폰켓몬 종류 수보다 내가 가져갈 수 있는 마리 수(`N // 2`)가 더 적다면, 아무리 골라도 가져갈 수 있는 마리 수만큼만 종류를 가져갈 수 있습니다.
  - 반대로 폰켓몬 종류 수가 더 적다면, 모든 종류를 다 챙겨가도 자리가 남으므로 종류 수만큼만 가져갈 수 있습니다.
  - 즉, **`N // 2`와 `중복을 제거한 폰켓몬 종류 수` 중 더 작은 값**을 반환하면 됩니다.

```python
def solution(nums):
    s = len(set(nums))
    tk = len(nums) // 2
    count = 0
    
    if s > tk:
        count = tk
    else:
        count = s
    
    return count
```
