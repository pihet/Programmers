###  완주하지 못한 선수 — 해시
참가자 명단과 완주자 명단이 주어진다. 완주하지 못한 사람 한 명의 이름을 반환

participant = ["leo", "kiki", "eden"]
completion = ["eden", "kiki"]

# 결과: "leo"

```python
not_in = ""
for i in participant:
    for j in completion:
        if i != j:
            not_in = i
print(f"참석하지 못한 선수는{not_in} 선수 입니다.")

#해당 방식으로 진행했을 때 동명이인 문제가 발생할 확률이 있고, 시간복잡도가 o(n^2)라는 아쉬운 상황
```

```python
counts = {}
for name in participant:
    if name not in counts:
        counts[name] = 1
    else: 
        counts[name] += 1
for name in completion:
        counts[name] -=1
for name in counts:
    if counts[name] == 1:
        print(f"완주하지 못 한 선수는 {name} 선수 입니다.")

#해당 문제식의 경우에는 동명이인이 발생한 경우를 해결할 수 있고, 시간복잡도가 o(n)이라 처리속도가 빠름  
```

### 가장 가까운 같은 글자 — 딕셔너리
문자열 s에서 각 글자가 이전에 등장한 적이 있으면, 현재 위치와 이전 위치의 차이를 넣어. 처음 나온 글자면 -1을 넣어.

s = "banana"

# 결과: [-1, -1, -1, 2, 2, 2]

```python
result = []
near = {}
for index, word in enumerate(s):
    if word not in near:
        result.append(-1)
    else:
        result.appned(index - near[word])
    near[word] = index

#enumerate 함수를 사용해서 글자의 처음과 이후의 인덱스 위치를 같이 추출해 비교를 함
```
