## 문제 1: 탐욕법(Greedy) - 큰 수 만들기 (Level 2)

- **문제 출처**: [프로그래머스 큰 수 만들기](https://school.programmers.co.kr/learn/courses/30/lessons/42883)

---

### 문제 설명

어떤 숫자에서 k개의 수를 제거했을 때 얻을 수 있는 가장 큰 숫자를 구하려 합니다.

예를 들어, 숫자 `1924`에서 수 두 개를 제거하면 `[19, 12, 14, 92, 94, 24]`를 만들 수 있습니다. 이 중 가장 큰 숫자는 `94`입니다.

문자열 형태의 숫자 `number`와 제거할 수의 개수 `k`가 매개변수로 주어질 때, `number`에서 `k` 개의 수를 제거했을 때 만들 수 있는 수 중 가장 큰 숫자를 문자열 형태로 return 하도록 solution 함수를 완성하세요.

---

### 입출력 예시 및 작동 흐름 설명

| number | k | return |
| --- | --- | --- |
| "1924" | 2 | "94" |
| "1231234" | 3 | "3234" |
| "4177252841" | 4 | "775841" |

#### 진행 과정 및 상세 작동 흐름 (number = "1924", k = 2 예시)
1. 첫 번째 숫자 `'1'`을 스택에 담습니다: `stack = ['1']`
2. 두 번째 숫자 `'9'`는 스택 상단 `'1'`보다 크므로, `'1'`을 제거하고 `'9'`를 넣습니다: `stack = ['9']`, `k = 1`
3. 세 번째 숫자 `'2'`를 넣습니다: `stack = ['9', '2']`
4. 네 번째 숫자 `'4'`는 스택 상단 `'2'`보다 크므로, `'2'`를 제거하고 `'4'`를 넣습니다: `stack = ['9', '4']`, `k = 0`
5. `k`개 제거가 완료되었으므로 최종 문자열 `"94"`를 반환합니다.


```python
def solution(number, k):
    stack = []
    for num in number:
        while stack and stack[-1] < num and k > 0:
            stack.pop()
            k -= 1
        stack.append(num)
    if k > 0:
        stack = stack[:-k]
    return "".join(stack)
```

## 문제 2: 스택/큐(Stack/Queue) - 같은 숫자는 싫어 (Level 1)

- **문제 출처**: [프로그래머스 같은 숫자는 싫어](https://school.programmers.co.kr/learn/courses/30/lessons/12906)

### 문제 설명

배열 `arr`가 주어집니다. 배열 `arr`의 각 원소는 숫자 0부터 9까지로 이루어져 있습니다. 이때, 배열 `arr`에서 연속적으로 나타나는 숫자는 하나만 남기고 전부 제거하려고 합니다. 단, 제거된 후 남은 수들을 반환할 때는 배열 `arr`의 원소 순서를 유지해야 합니다.

예를 들면,
- `arr = [1, 1, 3, 3, 0, 1, 1]` 이면 `[1, 3, 0, 1]` 을 return 합니다.
- `arr = [4, 4, 4, 3, 3]` 이면 `[4, 3]` 을 return 합니다.

배열 `arr`가 매개변수로 주어질 때, 제거하고 남은 수들을 return 하도록 solution 함수를 완성해 주세요.

---

### 입출력 예시 및 작동 흐름 설명

| arr | return |
| --- | --- |
| [1, 1, 3, 3, 0, 1, 1] | [1, 3, 0, 1] |
| [4, 4, 4, 3, 3] | [4, 3] |

#### 진행 과정 및 상세 작동 흐름
1. 결과를 담을 리스트 `answer`를 생성합니다.
2. `arr`의 각 원소를 순회하며, `answer`의 마지막 원소와 다를 경우에만 `answer`에 추가합니다.
3. 최종적으로 연속 중복이 제거된 배열을 반환합니다.


```python
def solution(arr):
    answer = []
    for i in arr:
        if not answer or answer[-1] != i:
            answer.append(i)
    return answer
```

## 문제 3: 스택/큐(Stack/Queue) - 기능개발 (Level 2)

- **문제 출처**: [프로그래머스 기능개발](https://school.programmers.co.kr/learn/courses/30/lessons/42586)

---

### 문제 설명

프로그래머스 팀에서는 기능 개선 작업을 수행 중입니다. 각 기능은 진도가 100%일 때 서비스에 반영할 수 있습니다.

또, 각 기능의 개발속도는 모두 다르기 때문에 뒤에 있는 기능이 앞에 있는 기능보다 먼저 개발될 수 있고, 이때 뒤에 있는 기능은 앞에 있는 기능이 배포될 때 함께 배포됩니다.

먼저 배포되어야 하는 순서대로 작업의 진도가 적힌 정수 배열 `progresses`와 각 작업의 개발 속도가 적힌 정수 배열 `speeds`가 주어질 때 각 배포마다 몇 개의 기능이 배포되는지를 return 하도록 solution 함수를 작성하세요.

---

### 입출력 예시 및 작동 흐름 설명

| progresses | speeds | return |
| --- | --- |
| [93, 30, 55] | [1, 30, 5] | [2, 1] |
| [95, 90, 99, 99, 80, 99] | [1, 1, 1, 1, 1, 1] | [1, 3, 2] |

#### 진행 과정 및 상세 작동 흐름 (progresses = [93, 30, 55], speeds = [1, 30, 5] 예시)
1. 각 작업 완수까지 필요한 남은 일수 계산:
   - 1번 작업(93%): (100 - 93) / 1 = 7일
   - 2번 작업(30%): (100 - 30) / 30 = 2.33.. -> 3일
   - 3번 작업(55%): (100 - 55) / 5 = 9일
2. 배포 그룹 묶기:
   - 1번 작업 배포일 7일에 2번 작업(3일 소요)도 함께 배포 가능 -> 첫 배포 2개
   - 3번 작업은 9일에 배포 -> 두 번째 배포 1개
3. 따라서 `[2, 1]`을 반환합니다.


```python
import math

def solution(progresses, speeds):
    a = []
    answer = []
    for i in range(len(progresses)):
        a.append(math.ceil((100 - progresses[i]) / speeds[i]))
    
    max_day = a[0]
    count = 1
    for i in a[1:]:
        if i <= max_day:
            count += 1
        else:
            answer.append(count)
            max_day = i
            count = 1
    answer.append(count)
    return answer
```
