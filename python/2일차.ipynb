### 가장 가까운 같은 글자 — 딕셔너리
문자열 s에서 각 글자가 이전에 등장한 적이 있으면, 현재 위치와 이전 위치의 차이를 넣어. 처음 나온 글자면 -1을 넣어.

s = "banana"

결과: [-1, -1, -1, 2, 2, 2]

```python
result = []
near = {}
for index, word in enumerate(s):
    if word not in near:
        result.append(-1)
    else:
        result.appned(index - near[word])
    near[word] = index

#enumerate 함수를 사용해서 글자의 처음과 이후의 인덱스 위치를 같이 추출해 비교를 함
```

### 괄호 검사 — 스택 입문
### 괄호 문자열이 올바른지 True/False로 반환해.

```python
s = "()()"

# 결과: True
s = "(())()"

# 결과: True
s = "())("

# 결과: False
```

```python
result = 0
for i in s:
    if i == "(" :
        result += 1
    else:
        result -= 1
        if result < 0:
            return False
            
if result == 0:
    return True
else:
    return False
```

### 문제: 연속된 같은 문자 제거하기
문자열에서 서로 붙어 있는 같은 문자 두 개를 제거해. 제거한 뒤에도 새로 같은 문자가 붙으면 계속 제거해. 모든 제거가 끝난 뒤 문자열이 비어 있으면 True, 남아 있으면 False를 반환해.

```python
s = "baabaa"
```

```python
stack = []
for i in s:
    if stack and stack[-1] == i:
        stack.pop()
    else:
        stack.append(i)
if not stack:
    return True
else:
    return False
```
