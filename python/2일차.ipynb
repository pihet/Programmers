### 가장 가까운 같은 글자 — 딕셔너리
문자열 s에서 각 글자가 이전에 등장한 적이 있으면, 현재 위치와 이전 위치의 차이를 넣어. 처음 나온 글자면 -1을 넣어.

s = "banana"

결과: [-1, -1, -1, 2, 2, 2]

```python
result = []
near = {}
for index, word in enumerate(s):
    if word not in near:
        result.append(-1)
    else:
        result.appned(index - near[word])
    near[word] = index

#enumerate 함수를 사용해서 글자의 처음과 이후의 인덱스 위치를 같이 추출해 비교를 함
```

### 괄호 검사 — 스택 입문
### 괄호 문자열이 올바른지 True/False로 반환해.

```python
s = "()()"

# 결과: True
s = "(())()"

# 결과: True
s = "())("

# 결과: False
```

```python
result = 0
for i in s:
    if i == "(" :
        result += 1
    else:
        result -= 1
        if result < 0:
            return False
            
if result == 0:
    return True
else:
    return False
```

### 문제: 연속된 같은 문자 제거하기
문자열에서 서로 붙어 있는 같은 문자 두 개를 제거해. 제거한 뒤에도 새로 같은 문자가 붙으면 계속 제거해. 모든 제거가 끝난 뒤 문자열이 비어 있으면 True, 남아 있으면 False를 반환해.

```python
s = "baabaa"
```

```python
def solution(s):
    stack = []
    for i in s:
        if stack and stack[-1] == i:
            stack.pop()
        else:
            stack.append(i)
    if not stack:
        return 1
    else:
        return 0
```

```python
"""
[프로그래머스 Lv.2 - 올바른 괄호]

- 문제 설명: 
  '(' 또는 ')' 로만 이루어진 문자열 s가 주어졌을 때, 
  괄호가 올바르게 짝지어졌으면 True, 올바르지 않으면 False를 return 하세요.

- 입출력 예시:
  s = "()()"   -> return True
  s = "(())()" -> return True
  s = ")()("   -> return False
  s = "(()("   -> return False

- 힌트:
  1. '(' 는 stack.append('(')
  2. ')' 일 때:
     - stack이 비어있으면 짝이 안 맞으므로 바로 return False
     - stack이 안 비어있으면 stack.pop()
  3. 반복문이 완전히 끝난 후:
     - stack이 비어있으면 return True (짝이 다 맞음)
     - stack에 '('가 남아있으면 return False
"""

def solution(s):
    stack = []
    for i in s:
        if i == "(":
            stack.append(i)
        else:
            if not stack:
                return False
            stack.pop()
    return not stack
```
