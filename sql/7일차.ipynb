# 7일차 SQL 학습 (Level 3 ~ 4)

## 문제 1: JOIN - 오랜 기간 보호한 동물(2) (Level 3)

- **문제 출처**: [프로그래머스 오랜 기간 보호한 동물(2)](https://school.programmers.co.kr/learn/courses/30/lessons/59411)

---

### 문제 설명

입양을 간 동물 중, 보호 기간이 가장 길었던 동물 두 마리의 ID와 이름을 조회하는 SQL문을 작성해주세요.
결과는 보호 기간이 긴 순으로 정렬해야 합니다.

---

### 예시 테이블 및 기대 출력 결과

#### ANIMAL_INS (입소 동물 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | INTAKE_CONDITION | NAME | SEX_UPON_INTAKE |
| --- | --- | --- | --- | --- | --- |
| A354597 | Cat | 2014-05-02 12:02:00 | Normal | Ariel | Spayed Female |
| A362707 | Dog | 2016-01-27 12:27:00 | Sick | Girly Girl | Spayed Female |
| A370507 | Cat | 2014-10-27 14:43:00 | Normal | Emily | Spayed Female |

#### ANIMAL_OUTS (입양 동물 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | NAME | SEX_UPON_OUTCOME |
| --- | --- | --- | --- | --- |
| A354597 | Cat | 2014-06-03 09:45:00 | Ariel | Spayed Female |
| A362707 | Dog | 2017-01-10 10:44:00 | Girly Girl | Spayed Female |

#### 기대하는 출력 결과 (RESULT)
| ANIMAL_ID | NAME |
| --- | --- |
| A362707 | Girly Girl |
| A354597 | Ariel |


```sql
-- 코드를 입력하세요
SELECT A.ANIMAL_ID, A.NAME
FROM ANIMAL_INS A JOIN ANIMAL_OUTS B ON A.ANIMAL_ID = B.ANIMAL_ID
ORDER BY DATEDIFF(B.DATETIME, A.DATETIME) DESC
LIMIT 2;
```

## 문제 2: GROUP BY - 식품분류별 가장 높은 가격의 식품 정보 조회하기 (Level 4)

- **문제 출처**: [프로그래머스 식품분류별 가장 높은 가격의 식품 정보 조회하기](https://school.programmers.co.kr/learn/courses/30/lessons/131116)

---

### 문제 설명

`FOOD_PRODUCT` 테이블에서 식품분류별로 가격이 가장 높은 식품의 분류, 가격, 이름을 조회하는 SQL문을 작성해주세요.
이때 식품분류가 '과자', '국', '김치', '식용유'인 경우만 출력하고, 결과는 식품 가격을 기준으로 내림차순 정렬해야 합니다.

---

### 예시 테이블 및 기대 출력 결과

#### FOOD_PRODUCT (식품 정보 테이블)
| PRODUCT_ID | PRODUCT_NAME | PRODUCT_CD | CATEGORY | PRICE |
| --- | --- | --- | --- | --- |
| P0018 | 맛있는마요네즈 | CD_OL101 | 식용유 | 5150 |
| P0019 | 고소한옥수수유 | CD_OL102 | 식용유 | 4500 |
| P0020 | 신라면 | CD_ND101 | 면 | 950 |
| P0021 | 맛있는김치 | CD_KC101 | 김치 | 19000 |
| P0022 | 독도곤드레김치 | CD_KC102 | 김치 | 12000 |

#### 기대하는 출력 결과 (RESULT)
| CATEGORY | MAX_PRICE | PRODUCT_NAME |
| --- | --- | --- |
| 김치 | 19000 | 맛있는김치 |
| 식용유 | 5150 | 맛있는마요네즈 |


```sql
-- 코드를 입력하세요
SELECT CATEGORY, PRICE AS MAX_PRICE, PRODUCT_NAME
FROM FOOD_PRODUCT
WHERE (CATEGORY, PRICE) IN (
    SELECT CATEGORY, MAX(PRICE)
    FROM FOOD_PRODUCT
    WHERE CATEGORY IN ('과자', '국', '김치', '식용유')
    GROUP BY CATEGORY
)
ORDER BY PRICE DESC;
```

## 문제 3: String, Date - 취소되지 않은 진료 예약 조회하기 (Level 4)

- **문제 출처**: [프로그래머스 취소되지 않은 진료 예약 조회하기](https://school.programmers.co.kr/learn/courses/30/lessons/132204)

---

### 문제 설명

`PATIENT`, `DOCTOR`, `APPOINTMENT` 테이블에서 2022년 4월 13일에 취소되지 않은 흉부외과(`CS`) 진료 예약 내역을 조회하는 SQL문을 작성해주세요.
진료예약번호(`APNT_NO`), 환자이름(`PT_NAME`), 환자번호(`PT_NO`), 진료과코드(`MCDP_CD`), 의사이름(`DR_NAME`), 진료예약일시(`APNT_YMD`)를 출력하고, 결과는 진료예약일시를 기준으로 오름차순 정렬해야 합니다.

---

### 예시 테이블 및 기대 출력 결과

#### PATIENT (환자 정보 예시 테이블)
| PT_NO | PT_NAME | GEND_CD | AGE | TLNO |
| --- | --- | --- | --- | --- |
| P20210001 | 라라 | F | 20 | 01011112222 |
| P20210002 | 루루 | M | 35 | 01033334444 |

#### DOCTOR (의사 정보 예시 테이블)
| DR_NAME | DR_ID | LCNS_NO | HIRE_YMD | MCDP_CD | TLNO |
| --- | --- | --- | --- | --- | --- |
| 이의사 | D2010001 | 12345 | 2010-03-01 | CS | 01055556666 |
| 김의사 | D2010002 | 12346 | 2011-04-01 | GS | 01077778888 |

#### APPOINTMENT (진료 예약 목록 예시 테이블)
| APNT_NO | PT_NO | MCDP_CD | MDDR_ID | APNT_YMD | APNT_CNCL_YN | APNT_CNCL_YMD |
| --- | --- | --- | --- | --- | --- | --- |
| 45 | P20210001 | CS | D2010001 | 2022-04-13 12:30:00 | N | NULL |
| 46 | P20210002 | CS | D2010001 | 2022-04-13 14:00:00 | Y | 2022-04-13 10:00:00 |
| 47 | P20210001 | GS | D2010002 | 2022-04-13 15:30:00 | N | NULL |

#### 기대하는 출력 결과 (RESULT)
| APNT_NO | PT_NAME | PT_NO | MCDP_CD | DR_NAME | APNT_YMD |
| --- | --- | --- | --- | --- | --- |
| 45 | 라라 | P20210001 | CS | 이의사 | 2022-04-13 12:30:00 |


```sql
-- 코드를 입력하세요
SELECT B.APNT_NO, A.PT_NAME, B.PT_NO, C.MCDP_CD, C.DR_NAME, B.APNT_YMD
FROM PATIENT A JOIN APPOINTMENT B ON A.PT_NO = B.PT_NO JOIN DOCTOR C ON B.MDDR_ID = C.DR_ID
WHERE DATE_FORMAT(B.APNT_YMD, "%Y-%m-%d") = "2022-04-13" AND B.APNT_CNCL_YN = "N" AND B.MCDP_CD = "CS"
ORDER BY B.APNT_YMD ASC;
```
