# 4일차 SQL 학습

## 문제 1: JOIN - 오랜 기간 보호한 동물(1) (Level 3)

- **문제 출처**: [프로그래머스 오랜 기간 보호한 동물(1)](https://school.programmers.co.kr/learn/courses/30/lessons/59044)

---

### 문제 설명

아직 입양을 못 간 동물 중, 가장 오래 보호소에 있었던 동물 3마리의 이름과 보호 시작일을 조회하는 SQL문을 작성해주세요. 이때 결과는 보호 시작일 순으로 조회해야 합니다.

---

### 예시 테이블 및 기대 출력 결과

#### ANIMAL_INS (보호소 입소 동물 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | INTAKE_CONDITION | NAME | SEX_UPON_INTAKE |
| --- | --- | --- | --- | --- | --- |
| A354597 | Cat | 2016-05-02 05:30:00 | Normal | Ariel | Spayed Female |
| A373687 | Dog | 2014-03-20 12:30:00 | Normal | Rosie | Spayed Female |
| A412697 | Dog | 2016-01-03 16:25:00 | Normal | Jackie | Neutered Male |
| A413789 | Dog | 2016-04-19 13:07:00 | Normal | Benji | Neutered Male |

#### ANIMAL_OUTS (보호소 퇴소/입양 동물 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | NAME | SEX_UPON_OUTCOME |
| --- | --- | --- | --- | --- |
| A354597 | Cat | 2016-06-02 09:49:00 | Ariel | Spayed Female |
| A373687 | Dog | 2014-03-20 14:26:00 | Rosie | Spayed Female |

#### 기대하는 출력 결과 (RESULT)
| NAME | DATETIME |
| --- | --- |
| Jackie | 2016-01-03 16:25:00 |
| Benji | 2016-04-19 13:07:00 |

---

### 입출력 예시 및 작동 흐름 설명

1. ANIMAL_INS 테이블에는 4마리(Ariel, Rosie, Jackie, Benji)가 있습니다.
2. ANIMAL_OUTS 테이블에는 2마리(Ariel, Rosie)만 입양 기록이 있습니다.
3. 입양을 못 간 동물은 ANIMAL_OUTS에 조인되지 않은 동물인 `Jackie`와 `Benji`입니다.
4. 이 중 보호 시작일(DATETIME)이 더 오래된 순서대로 정렬하여 상위 레코드를 출력합니다.

---

### 단계별 생각 유도 힌트 및 의사코드 (Pseudocode)

#### 생각 유도 힌트
1. **LEFT JOIN의 활용**: `ANIMAL_INS`를 기준(LEFT)으로 `ANIMAL_OUTS`와 `ANIMAL_ID`로 조인하면, 입양을 못 간 동물은 `ANIMAL_OUTS.ANIMAL_ID` 값이 `NULL`이 됩니다.
2. **입양 미완료 조건**: `WHERE ANIMAL_OUTS.ANIMAL_ID IS NULL` 조건으로 입양되지 않은 동물만 필터링합니다.
3. **정렬 및 개수 제한**: 가장 오래 보호소에 있었던 순서(보호 시작일 오름차순 `DATETIME ASC`)로 정렬한 후 상위 3개(`LIMIT 3`)만 조회합니다.

#### 의사코드 (Pseudocode)
```text
SELECT ANIMAL_INS의 NAME과 DATETIME
FROM ANIMAL_INS
LEFT JOIN ANIMAL_OUTS ON ANIMAL_INS.ANIMAL_ID = ANIMAL_OUTS.ANIMAL_ID
WHERE ANIMAL_OUTS.ANIMAL_ID IS NULL
ORDER BY ANIMAL_INS.DATETIME 오름차순
LIMIT 3;
```
