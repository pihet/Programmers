# 10일차 SQL 알고리즘 학습

## 문제 1: 우유와 요거트가 담긴 바구니 (Level 4)

- **문제 출처**: [프로그래머스 우유와 요거트가 담긴 바구니](https://school.programmers.co.kr/learn/courses/30/lessons/62284)

---

### 문제 설명

`CART_PRODUCTS` 테이블은 장바구니에 담긴 상품 정보를 담은 테이블입니다. 우유('Milk')와 요거트('Yogurt')를 동시에 구입한 장바구니의 아이디를 조회하는 SQL 문을 작성해주세요. 이때 결과는 장바구니의 아이디 순으로 정렬해야 합니다.

---

### 예시 테이블 및 기대 결과

#### CART_PRODUCTS 테이블 예시
| ID | CART_ID | NAME | PRICE |
| --- | --- | --- | --- |
| 8834 | 286 | Milk | 2540 |
| 8835 | 286 | Yogurt | 1500 |
| 8836 | 448 | Milk | 2540 |
| 8837 | 1034 | Yogurt | 1500 |

#### 기대하는 출력 결과 (RESULT)
| CART_ID |
| --- |
| 286 |

#### 상세 작동 흐름
1. 286번 장바구니는 Milk와 Yogurt를 모두 담고 있으므로 출력 대상입니다.
2. 448번 장바구니는 Milk만, 1034번 장바구니는 Yogurt만 담고 있으므로 제외됩니다.
3. 따라서 286번 CART_ID가 정렬되어 출력됩니다.

```sql
-- 코드를 작성해주세요.
SELECT CART_ID
FROM CART_PRODCUTS 
WHERE NAME IN ('Milk','Yogurt')
GROUP BY CART_ID
HAVING COUNT(DISTINCT NAME) = 2
ORDER BY CART_ID ASC;
```

## 문제 2: 없어진 기록 찾기 (Level 3)

- **문제 출처**: [프로그래머스 없어진 기록 찾기](https://school.programmers.co.kr/learn/courses/30/lessons/59042)

---

### 문제 설명

천재지변으로 인해 일부 데이터가 유실되었습니다. 입양을 간 기록은 있는데, 보호소에 들어온 기록이 없는 동물의 ID와 이름을 ANIMAL_ID 순으로 조회하는 SQL문을 작성해주세요.

---

### 예시 테이블 및 기대 결과

#### ANIMAL_INS (입고 기록)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | IN_CONDITION | NAME | SEX_UPON_INTAKE |
| --- | --- | --- | --- | --- | --- |
| A352713 | Dog | 2017-04-13 16:29:00 | Normal | Merida | Neutered Male |
| A350375 | Cat | 2017-03-06 15:01:00 | Normal | Meo | Spayed Female |

#### ANIMAL_OUTS (출고/입양 기록)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | NAME | SEX_UPON_OUTCOME |
| --- | --- | --- | --- | --- |
| A349733 | Dog | 2017-09-27 19:09:00 | Allie | Spayed Female |
| A352713 | Dog | 2017-04-25 12:25:00 | Merida | Neutered Male |

#### 기대하는 출력 결과 (RESULT)
| ANIMAL_ID | NAME |
| --- | --- |
| A349733 | Allie |

#### 상세 작동 흐름
1. Allie(A349733)는 입양 기록(ANIMAL_OUTS)에는 존재하지만, 입고 기록(ANIMAL_INS)에는 없습니다.
2. Merida(A352713)는 두 테이블에 모두 존재하므로 유실된 데이터가 아닙니다.
3. 따라서 유실된 Allie(A349733)의 ID와 이름을 ANIMAL_ID 오름차순으로 출력합니다.

```sql
-- 코드를 작성해주세요.
SELECT A.ANIMAL_ID, A.NAME
FROM ANIMAL_OUTS A LEFT JOIN ANIMAL_INS B ON A.ANIMAL_ID = B.ANIMAL_ID
WHERE A.ANIMAL_ID IS NULL
ORDER BY A.ANIMAL_ID ASC
```

## 문제 3: 5월 식품들의 총매출 조회하기 (Level 4)

- **문제 출처**: [프로그래머스 5월 식품들의 총매출 조회하기](https://school.programmers.co.kr/learn/courses/30/lessons/131117)

---

### 문제 설명

`FOOD_PRODUCT`와 `FOOD_ORDER` 테이블에서 생산일자가 2022년 5월인 식품들의 식품 ID, 식품 이름, 총매출을 조회하는 SQL문을 작성해주세요. 이때 결과는 총매출을 기준으로 내림차순 정렬해주시고 총매출이 같다면 식품 ID를 기준으로 오름차순 정렬해주세요.

---

### 예시 테이블 및 기대 결과

#### FOOD_PRODUCT (식품 정보)
| PRODUCT_ID | PRODUCT_NAME | PRODUCT_CD | CATEGORY | PRICE |
| --- | --- | --- | --- | --- |
| P0015 | 맛있는배추김치 | CD_KC00001 | 김치 | 19000 |
| P0016 | 맛있는쌀 | CD_KC00002 | 국 | 24000 |

#### FOOD_ORDER (주문 정보)
| ORDER_ID | PRODUCT_ID | AMOUNT | PRODUCE_DATE | IN_DATE | OUT_DATE | FACTORY_ID | ORDER_STATUS |
| --- | --- | --- | --- | --- | --- | --- | --- |
| O0001 | P0015 | 10 | 2022-05-15 | 2022-05-16 | 2022-05-17 | FT0001 | COMPLETE |
| O0002 | P0015 | 5 | 2022-05-20 | 2022-05-21 | 2022-05-22 | FT0001 | COMPLETE |
| O0003 | P0016 | 3 | 2022-04-10 | 2022-04-11 | 2022-04-12 | FT0002 | COMPLETE |

#### 기대하는 출력 결과 (RESULT)
| PRODUCT_ID | PRODUCT_NAME | TOTAL_SALES |
| --- | --- | --- |
| P0015 | 맛있는배추김치 | 285000 |

#### 상세 작동 흐름
1. 2022년 5월에 생산된 P0015 상품의 주문 수량 합계는 10 + 5 = 15개입니다.
2. 단가 19,000원 * 15개 = 총매출 285,000원이 됩니다.
3. P0016은 2022년 4월 생산이므로 대상에서 제외됩니다.

```sql
-- 코드를 작성해주세요.
SELECT B.PRODUCT_ID, B.PRODUCT_NAME, SUM(B.PRICE * A.AMOUNT) AS TOTAL_SLAES
FROM FOOD_ORDER A LEFT JOIN FOOD_PRODUCT B ON A.PRODUCT_ID = B.PRODUCT_ID
WHERE PRODUCE_DATE  LIKE '2022-05%'
GROUP BY B.PRODUCT_ID, B.PRODUCT_NAME
ORDER BY TOTAL_SLAES DESC, B.PRODUCT_ID ASC;
```
