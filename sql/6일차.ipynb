# 6일차 SQL 학습

## 문제 1: JOIN - 오랜 기간 보호한 동물(1) (Level 3)

- **문제 출처**: [프로그래머스 오랜 기간 보호한 동물(1)](https://school.programmers.co.kr/learn/courses/30/lessons/59044)

---

### 문제 설명

아직 입양을 못 간 동물 중, 가장 오래 보호소에 있었던 동물 3마리의 이름과 입소일을 조회하는 SQL문을 작성해주세요. 이때 결과는 입소일 순으로 조회해야 합니다.

---

### 예시 테이블 및 기대 출력 결과

#### ANIMAL_INS (보호소 입소 동물 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | INTAKE_CONDITION | NAME | SEX_UPON_INTAKE |
| --- | --- | --- | --- | --- | --- |
| A354597 | Cat | 2014-05-02 12:16:00 | Normal | Ariel | Spayed Female |
| A373687 | Dog | 2014-03-20 12:31:00 | Normal | Rosie | Normal |
| A362707 | Dog | 2016-01-27 12:27:00 | Normal | Girly Girl | Spayed Female |
| A412697 | Dog | 2016-01-03 16:25:00 | Normal | Jackie | Neutered Male |

#### ANIMAL_OUTS (보호소 퇴소/입양 동물 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | NAME | SEX_UPON_OUTCOME |
| --- | --- | --- | --- | --- |
| A354597 | Cat | 2014-05-06 14:44:00 | Ariel | Spayed Female |

#### 기대하는 출력 결과 (RESULT)
| NAME | DATETIME |
| --- | --- |
| Rosie | 2014-03-20 12:31:00 |
| Jackie | 2016-01-03 16:25:00 |
| Girly Girl | 2016-01-27 12:27:00 |

---

### 입출력 예시 및 작동 흐름 설명

1. `ANIMAL_INS`에는 Ariel, Rosie, Girly Girl, Jackie 4마리가 있습니다.
2. `ANIMAL_OUTS`에는 Ariel만 존재하므로, 입양을 못 간 동물은 Rosie, Girly Girl, Jackie 3마리입니다.
3. 입소일(`DATETIME`) 오름차순으로 정렬하면 Rosie(2014년), Jackie(2016년 1월 3일), Girly Girl(2016년 1월 27일) 순입니다.
4. 상위 3마리의 이름과 입소일을 조회합니다.


```sql
-- 코드를 입력하세요
SELECT A.NAME, A.DATETIME
FROM ANIMAL_INS A LEFT JOIN ANIMAL_OUTS B ON A.ANIMAL_ID = B.ANIMAL_ID
WHERE B.ANIMAL_ID IS NULL
ORDER BY A.DATETIME ASC
LIMIT 3;
```

## 문제 2: GROUP BY - 즐겨찾기가 가장 많은 식당 정보 출력하기 (Level 3)

- **문제 출처**: [프로그래머스 즐겨찾기가 가장 많은 식당 정보 출력하기](https://school.programmers.co.kr/learn/courses/30/lessons/131123)

---

### 문제 설명

`REST_INFO` 테이블에서 음식종류별로 즐겨찾기수가 가장 많은 식당의 음식 종류, ID, 식당 이름, 즐겨찾기수를 조회하는 SQL문을 작성해주세요. 이때 결과는 음식 종류를 기준으로 내림차순 정렬해주세요.

---

### 예시 테이블 및 기대 출력 결과

#### REST_INFO (식당 정보 테이블)
| REST_ID | REST_NAME | FOOD_TYPE | VIEWS | FAVORITES | PARKING_LOT | ADDRESS | TEL |
| --- | --- | --- | --- | --- | --- | --- | --- |
| 00001 | 은하식당 | 한식 | 10000 | 734 | N | 서울시 ... | 02-000-0000 |
| 00002 | 하하밥집 | 한식 | 20000 | 1024 | Y | 서울시 ... | 02-111-1111 |
| 00003 | 맛있는일식 | 일식 | 8000 | 500 | Y | 서울시 ... | 02-222-2222 |
| 00004 | 스시나라 | 일식 | 15000 | 1500 | N | 서울시 ... | 02-333-3333 |

#### 기대하는 출력 결과 (RESULT)
| FOOD_TYPE | REST_ID | REST_NAME | FAVORITES |
| --- | --- | --- | --- |
| 한식 | 00002 | 하하밥집 | 1024 |
| 일식 | 00004 | 스시나라 | 1500 |

---

### 입출력 예시 및 작동 흐름 설명

1. 한식 카테고리에서 즐겨찾기수가 가장 많은 식당은 하하밥집(1024개)입니다.
2. 일식 카테고리에서 즐겨찾기수가 가장 많은 식당은 스시나라(1500개)입니다.
3. 각 음식종류별 최대 즐겨찾기를 보유한 식당의 정보를 추출하고 음식종류(`FOOD_TYPE`) 내림차순으로 정렬합니다.


```sql
-- 코드를 입력하세요
SELECT FOOD_TYPE, REST_ID, REST_NAME, FAVORITES
FROM REST_INFO
WHERE (FOOD_TYPE,FAVORITES) IN (SELECT FOOD_TYPE, MAX(FAVORITES) FROM REST_INFO GROUP BY FOOD_TYPE)
ORDER BY FOOD_TYPE DESC;
```
