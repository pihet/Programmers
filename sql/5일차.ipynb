# 5일차 SQL 학습

## 문제 1: JOIN - 없어진 기록 찾기 (Level 3)

- **문제 출처**: [프로그래머스 없어진 기록 찾기](https://school.programmers.co.kr/learn/courses/30/lessons/59042)

---

### 문제 설명

천재지변으로 인해 일부 데이터가 유실되었습니다. 입양을 간 기록은 있는데, 보호소에 들어온 기록이 없는 동물의 ID와 이름을 ID 순으로 조회하는 SQL문을 작성해주세요.

---

### 예시 테이블 및 기대 출력 결과

#### ANIMAL_INS (보호소 입소 동물 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | INTAKE_CONDITION | NAME | SEX_UPON_INTAKE |
| --- | --- | --- | --- | --- | --- |
| A352713 | Dog | 2017-04-13 16:29:00 | Normal | Gia | Spayed Female |
| A350375 | Cat | 2017-03-06 15:01:00 | Normal | Meow | Neutered Male |

#### ANIMAL_OUTS (보호소 퇴소/입양 동물 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | NAME | SEX_UPON_OUTCOME |
| --- | --- | --- | --- | --- |
| A349733 | Dog | 2017-09-27 19:09:00 | Allie | Spayed Female |
| A352713 | Dog | 2017-04-25 12:25:00 | Gia | Spayed Female |
| A349990 | Cat | 2018-02-02 14:18:00 | Spice | Spayed Female |

#### 기대하는 출력 결과 (RESULT)
| ANIMAL_ID | NAME |
| --- | --- |
| A349733 | Allie |
| A349990 | Spice |

---

### 입출력 예시 및 작동 흐름 설명

1. `ANIMAL_OUTS`에는 Allie, Gia, Spice가 있습니다.
2. `ANIMAL_INS`에는 Gia와 Meow만 기록되어 있습니다.
3. 따라서 입양 기록(`ANIMAL_OUTS`)은 존재하지만 입소 기록(`ANIMAL_INS`)이 유실된 동물은 `Allie`와 `Spice`입니다.
4. 이 두 동물의 ID와 이름을 `ANIMAL_ID` 오름차순으로 조회합니다.

---

### 단계별 생각 유도 힌트 및 의사코드 (Pseudocode)

#### 생각 유도 힌트
1. **오른쪽 기준 LEFT / RIGHT JOIN**: 이번에는 `ANIMAL_OUTS`를 기준(LEFT)으로 `ANIMAL_INS`를 조인하거나, `ANIMAL_OUTS`에 `RIGHT JOIN`을 걸어 `ANIMAL_INS` 쪽 데이터가 존재하지 않는 행을 찾아야 합니다.
2. **유실 데이터 조건**: `WHERE ANIMAL_INS.ANIMAL_ID IS NULL` 조건을 적용하여 입소 기록이 없는 동물을 필터링합니다.
3. **정렬**: `ORDER BY ANIMAL_OUTS.ANIMAL_ID ASC`로 정렬합니다.

#### 의사코드 (Pseudocode)
```text
SELECT ANIMAL_OUTS의 ANIMAL_ID, NAME
FROM ANIMAL_OUTS
LEFT JOIN ANIMAL_INS ON ANIMAL_OUTS.ANIMAL_ID = ANIMAL_INS.ANIMAL_ID
WHERE ANIMAL_INS.ANIMAL_ID IS NULL
ORDER BY ANIMAL_OUTS.ANIMAL_ID 오름차순;
```

## 문제 2: JOIN - 보호소에서 중성화한 동물 (Level 3)

- **문제 출처**: [프로그래머스 보호소에서 중성화한 동물](https://school.programmers.co.kr/learn/courses/30/lessons/59045)

---

### 문제 설명

보호소에서 중성화 수술을 거친 동물 정보를 알아보려 합니다. 보호소에 들어올 당시에는 중성화되지 않았지만, 보호소를 나갈 당시에는 중성화된 동물의 아이디와 생물 종, 이름을 아이디 순으로 조회하는 SQL문을 작성해주세요.

*(참고: 중성화되지 않은 동물 상태는 `Intact`, 중성화된 동물 상태는 `Spayed` 또는 `Neutered` 표기)*

---

### 예시 테이블 및 기대 출력 결과

#### ANIMAL_INS (보호소 입소 동물 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | INTAKE_CONDITION | NAME | SEX_UPON_INTAKE |
| --- | --- | --- | --- | --- | --- |
| A367438 | Dog | 2015-01-27 16:02:00 | Normal | Cookie | Intact Female |
| A382192 | Dog | 2015-03-13 08:47:00 | Normal | Element | Neutered Male |
| A410668 | Cat | 2015-11-19 13:41:00 | Normal | Raven | Intact Female |

#### ANIMAL_OUTS (보호소 퇴소/입양 동물 정보)
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | NAME | SEX_UPON_OUTCOME |
| --- | --- | --- | --- | --- |
| A367438 | Dog | 2015-01-29 14:12:00 | Cookie | Spayed Female |
| A382192 | Dog | 2015-03-16 13:14:00 | Element | Neutered Male |
| A410668 | Cat | 2015-11-19 14:10:00 | Raven | Spayed Female |

#### 기대하는 출력 결과 (RESULT)
| ANIMAL_ID | ANIMAL_TYPE | NAME |
| --- | --- | --- |
| A367438 | Dog | Cookie |
| A410668 | Cat | Raven |

---

### 입출력 예시 및 작동 흐름 설명

1. `Cookie`는 입소 시 `Intact Female`이었으나 퇴소 시 `Spayed Female`로 변경 (중성화 됨)
2. `Element`는 입소 시 이미 `Neutered Male`이었음 (해당 없음)
3. `Raven`은 입소 시 `Intact Female`이었으나 퇴소 시 `Spayed Female`로 변경 (중성화 됨)
4. 따라서 결과는 Cookie와 Raven 2마리가 아이디 순으로 조회됩니다.

---

### 단계별 생각 유도 힌트 및 의사코드 (Pseudocode)

#### 생각 유도 힌트
1. **조인 연결**: `ANIMAL_INS`와 `ANIMAL_OUTS`를 `ANIMAL_ID`로 `JOIN`합니다.
2. **입소 시 조건**: `ANIMAL_INS.SEX_UPON_INTAKE LIKE 'Intact%'` (중성화 전)
3. **퇴소 시 조건**: `ANIMAL_OUTS.SEX_UPON_OUTCOME LIKE 'Spayed%'` OR `ANIMAL_OUTS.SEX_UPON_OUTCOME LIKE 'Neutered%'` (중성화 후)
4. **정렬**: `ORDER BY ANIMAL_INS.ANIMAL_ID` 오름차순

#### 의사코드 (Pseudocode)
```text
SELECT A.ANIMAL_ID, A.ANIMAL_TYPE, A.NAME
FROM ANIMAL_INS A
JOIN ANIMAL_OUTS B ON A.ANIMAL_ID = B.ANIMAL_ID
WHERE A.SEX_UPON_INTAKE LIKE 'Intact%'
  AND (B.SEX_UPON_OUTCOME LIKE 'Spayed%' OR B.SEX_UPON_OUTCOME LIKE 'Neutered%')
ORDER BY A.ANIMAL_ID;
```
