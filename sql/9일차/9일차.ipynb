# 9일차 SQL 학습 (Level 3 ~ 4)

## 문제 1: GROUP BY / HAVING / 서브쿼리 - 헤비 유저가 소유한 장소 (Level 3)

- **문제 출처**: [프로그래머스 헤비 유저가 소유한 장소](https://school.programmers.co.kr/learn/courses/30/lessons/77487)

---

### 문제 설명

이 서비스에서는 공간을 2개 이상 등록한 사람을 "헤비 유저"라고 부릅니다. 헤비 유저가 등록한 공간의 정보를 아이디 순으로 조회하는 SQL문을 작성해주세요.

---

### 예시 테이블 및 기대 출력 결과

#### PLACES (공간 정보 테이블)
| ID | NAME | HOST_ID |
| --- | --- | --- |
| 14354 | 유니크한 연습실 | 760849 |
| 22692 | 럭셔리 스튜디오 | 760849 |
| 34901 | 공유 작업실 | 309400 |
| 45123 | 아늑한 회의실 | 760849 |

#### 기대하는 출력 결과 (RESULT - ID 순 정렬)
| ID | NAME | HOST_ID |
| --- | --- | --- |
| 14354 | 유니크한 연습실 | 760849 |
| 22692 | 럭셔리 스튜디오 | 760849 |
| 45123 | 아늑한 회의실 | 760849 |

---

### 입출력 예시 및 작동 흐름 설명
1. `HOST_ID`가 `760849`인 유저는 등록한 공간이 3개(14354, 22692, 45123)이므로 헤비 유저에 해당합니다.
2. `HOST_ID`가 `309400`인 유저는 공간을 1개만 등록했으므로 제외됩니다.
3. 따라서 헤비 유저 `760849`의 공간 3개가 ID 오름차순으로 출력됩니다.

```sql
-- 코드를 입력하세요
SELECT ID, NAME, HOST_ID
FROM PLACES
WHERE HOST_ID IN (SELECT HOST_ID FROM PLACES GROUP BY HOST_ID HAVING COUNT(HOST_ID) >= 2)
ORDER BY ID ASC;
```

## 문제 2: UNION ALL / DATE_FORMAT - 오프라인/온라인 리얼타임 상품 판매 데이터 통합하기 (Level 4)

- **문제 출처**: [프로그래머스 오프라인/온라인 리얼타임 상품 판매 데이터 통합하기](https://school.programmers.co.kr/learn/courses/30/lessons/131537)

---

### 문제 설명

`ONLINE_SALE` 테이블과 `OFFLINE_SALE` 테이블에서 2022년 3월의 오프라인/온라인 상품 판매 데이터의 판매 날짜, 상품ID, 유저ID, 판매량을 출력하는 SQL문을 작성해주세요.
`OFFLINE_SALE` 테이블의 판매 데이터의 `USER_ID` 값은 `NULL`로 표시해주세요.
결과는 판매 날짜를 기준으로 오름차순 정렬해주시고 판매 날짜가 같다면 상품 ID를 기준으로 오름차순, 상품ID까지 같다면 유저 ID를 기준으로 오름차순 정렬해주세요.

---

### 예시 테이블 및 기대 출력 결과

#### ONLINE_SALE (온라인 상품 판매 테이블)
| ONLINE_SALE_ID | USER_ID | PRODUCT_ID | SALES_AMOUNT | SALES_DATE |
| --- | --- | --- | --- | --- |
| 1 | 1 | 22 | 1 | 2022-03-01 |
| 2 | 2 | 14 | 1 | 2022-03-02 |

#### OFFLINE_SALE (오프라인 상품 판매 테이블)
| OFFLINE_SALE_ID | PRODUCT_ID | SALES_AMOUNT | SALES_DATE |
| --- | --- | --- | --- | --- |
| 1 | 22 | 2 | 2022-03-01 |
| 2 | 14 | 1 | 2022-03-05 |

#### 기대하는 출력 결과 (RESULT)
| SALES_DATE | PRODUCT_ID | USER_ID | SALES_AMOUNT |
| --- | --- | --- | --- |
| 2022-03-01 | 22 | 1 | 1 |
| 2022-03-01 | 22 | NULL | 2 |
| 2022-03-02 | 14 | 2 | 1 |
| 2022-03-05 | 14 | NULL | 1 |

---

### 입출력 예시 및 작동 흐름 설명
1. 2022년 3월 데이터만 두 테이블에서 각각 추출합니다 (`SALES_DATE LIKE '2022-03%'` 또는 `DATE_FORMAT`).
2. `OFFLINE_SALE`에는 `USER_ID` 컬럼이 없으므로 `NULL AS USER_ID`로 맞춘 후 두 결과를 `UNION ALL`로 통합합니다.
3. 정렬 조건(날짜 ASC, 상품ID ASC, 유저ID ASC)에 따라 결과를 정렬합니다.

```sql
-- 코드를 입력하세요
SELECT DATE_FORMAT(SALES_DATE, '%Y-%m-%d') AS SALES_DATE, PRODUCT_ID, USER_ID, SALES_AMOUNT
FROM ONLINE_SALE
WHERE DATE_FORMAT(SALES_DATE, '%Y-%m') = '2022-03'
UNION ALL
SELECT DATE_FORMAT(SALES_DATE, '%Y-%m-%d') AS SALES_DATE, PRODUCT_ID, NULL AS USER_ID, SALES_AMOUNT
FROM OFFLINE_SALE
WHERE DATE_FORMAT(SALES_DATE, '%Y-%m') = '2022-03'
ORDER BY SALES_DATE ASC, PRODUCT_ID ASC, USER_ID ASC;
```
