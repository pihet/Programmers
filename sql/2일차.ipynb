# 2일차 SQL 학습

## [문제 1]: [프로그래머스 Lv.2 - 조건에 맞는 도서와 저자 리스트 출력하기](https://school.programmers.co.kr/learn/courses/30/lessons/144854)

### 문제 설명
`BOOK` 테이블과 `AUTHOR` 테이블에서 '경제' 카테고리에 속하는 도서들의 도서 ID(`BOOK_ID`), 저자명(`AUTHOR_NAME`), 출판일(`PUBLISHED_DATE`) 리스트를 출력하는 SQL문을 작성해주세요.
결과는 출판일을 기준으로 오름차순 정렬해주세요.

---

#### 예시 테이블 데이터

##### BOOK
| BOOK_ID | CATEGORY | AUTHOR_ID | PRICE | PUBLISHED_DATE |
| :--- | :--- | :--- | :--- | :--- |
| 1 | 인문 | 1 | 10000 | 2020-01-01 |
| 2 | 경제 | 1 | 9000 | 2021-04-11 |
| 3 | 경제 | 2 | 11000 | 2021-02-05 |

##### AUTHOR
| AUTHOR_ID | AUTHOR_NAME |
| :--- | :--- |
| 1 | 홍길동 |
| 2 | 김영호 |

##### 출력 예시 결과 (Expected Output)
| BOOK_ID | AUTHOR_NAME | PUBLISHED_DATE |
| :--- | :--- | :--- |
| 3 | 김영호 | 2021-02-05 |
| 2 | 홍길동 | 2021-04-11 |

---

#### 과거 기록 기반 주의할 문제점 복습 노트
- **날짜 포맷팅**: `PUBLISHED_DATE`의 출력 포맷을 `YYYY-MM-DD` 형식으로 맞추기 위해 `DATE_FORMAT(PUBLISHED_DATE, '%Y-%m-%d')`을 사용해야 합니다. 시/분/초가 포함되지 않도록 주의하세요.

#### 힌트 & 생각 유도 가이드
- `BOOK` 테이블과 `AUTHOR` 테이블을 공통 컬럼인 `AUTHOR_ID`를 이용해 `JOIN`합니다.
- `WHERE` 조건을 통해 `CATEGORY`가 '경제'인 데이터만 필터링합니다.
- 필요한 컬럼(`BOOK_ID`, `AUTHOR_NAME`, 날짜 포맷팅을 적용한 `PUBLISHED_DATE`)을 `SELECT`합니다.
- `ORDER BY`를 사용해 `PUBLISHED_DATE` 기준 오름차순으로 정렬합니다.

```sql
-- 코드를 입력하세요
SELECT BOOK_ID,  AUTHOR_NAME, DATE_FORMAT(PUBLISHED_DATE, "%Y-%m-%d") AS PUBLISHED_DATE
FROM BOOK A JOIN AUTHOR B ON A.AUTHOR_ID = B.AUTHOR_ID
WHERE A.CATEGORY = "경제"
ORDER BY A.PUBLISHED_DATE ASC;
```

## [문제 2]: [프로그래머스 Lv.2 - 성분으로 구분한 아이스크림 총 주문량](https://school.programmers.co.kr/learn/courses/30/lessons/133026)

### 문제 설명
`FIRST_HALF` 테이블과 `ICECREAM_INFO` 테이블에서 상반기 동안 각 아이스크림 성분 타입(`INGREDIENT_TYPE`)과 성분 타입별 아이스크림의 총주문량(`TOTAL_ORDER`)을 총주문량이 작은 순서대로 조회하는 SQL문을 작성해주세요. 이때 총주문량의 컬럼명은 `TOTAL_ORDER`로 지정해주세요.

---

#### 예시 테이블 데이터

##### FIRST_HALF
| SHIPMENT_ID | FLAVOR | TOTAL_ORDER |
| :--- | :--- | :--- |
| 101 | chocolate | 3200 |
| 102 | vanilla | 2800 |
| 103 | mint_chocolate | 1700 |
| 106 | peach | 2450 |

##### ICECREAM_INFO
| FLAVOR | INGREDIENT_TYPE |
| :--- | :--- |
| chocolate | sugar_based |
| vanilla | sugar_based |
| mint_chocolate | sugar_based |
| peach | fruit_based |

##### 출력 예시 결과 (Expected Output)
| INGREDIENT_TYPE | TOTAL_ORDER |
| :--- | :--- |
| fruit_based | 2450 |
| sugar_based | 7700 |

---

#### 과거 기록 기반 주의할 문제점 복습 노트
- **집계 함수와 GROUP BY**: 성분 타입별 총주문량을 구해야 하므로 `SUM(TOTAL_ORDER)`과 `GROUP BY INGREDIENT_TYPE`을 사용해야 합니다.
- **별칭 지정**: 최종 출력 결과의 컬럼명을 `TOTAL_ORDER`로 매핑하기 위해 `AS TOTAL_ORDER` 별칭을 지정해야 합니다.

#### 힌트 & 생각 유도 가이드
- `FIRST_HALF` 테이블과 `ICECREAM_INFO` 테이블을 공통 컬럼인 `FLAVOR`를 사용하여 `JOIN`합니다.
- `GROUP BY INGREDIENT_TYPE`을 통해 성분 타입별로 그룹화합니다.
- `SUM(TOTAL_ORDER)`으로 성분별 총주문량 합계를 구합니다.
- `ORDER BY`를 활용하여 총주문량 기준 오름차순 정렬을 수행합니다.

```sql
-- 코드를 입력하세요
SELECT B.INGREDIENT_TYPE, SUM(A.TOTAL_ORDER) AS TOTAL_ORDER
FROM FIRST_HALF A JOIN ICECREAM_INFO B ON A.FLAVOR = B.FLAVOR
GROUP BY B.INGREDIENT_TYPE
ORDER BY A.TOTAL_ORDER ASC;
```
