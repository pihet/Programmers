## 문제 1: JOIN / GROUP BY - 주문량이 많은 아이스크림들 조회하기 (Level 4)

- **문제 출처**: [프로그래머스 주문량이 많은 아이스크림들 조회하기](https://school.programmers.co.kr/learn/courses/30/lessons/133027)

---

### 문제 설명

7월 아이스크림 총 주문량과 상반기의 아이스크림 총 주문량을 더한 값이 큰 순서대로 상위 3개의 맛을 조회하는 SQL문을 작성해주세요.

---

### 예시 테이블 및 기대 출력 결과

#### FIRST_HALF (상반기 아이스크림 주문 정보)
| SHIPMENT_ID | FLAVOR | TOTAL_ORDER |
| --- | --- | --- |
| 101 | chocolate | 3200 |
| 102 | vanilla | 2800 |
| 103 | mint_chocolate | 1700 |
| 104 | caramel | 2600 |

#### JULY (7월 아이스크림 주문 정보 - 같은 맛도 출하 번호가 다를 수 있음)
| SHIPMENT_ID | FLAVOR | TOTAL_ORDER |
| --- | --- | --- |
| 201 | chocolate | 500 |
| 202 | chocolate | 700 |
| 203 | vanilla | 500 |
| 204 | mint_chocolate | 1000 |

#### 기대하는 출력 결과 (RESULT)
| FLAVOR |
| --- |
| chocolate |
| vanilla |
| mint_chocolate |


```python
-- 코드를 입력하세요
SELECT A.FLAVOR
FROM FIRST_HALF A (SELECT FLAVOR, SUM(TOTAL_ORDER) AS JULY_TOTAL FROM JULY GROUP BY FLAVOR) B ON A.FLAVOR = B.FALVOR
ORDER BY (A.TOTAL_ORDER+ B.JULY_TOTAL) DESC
LIMIT 3;
```

## 문제 2: String, Date / GROUP BY - 카테고리 별 도서 판매량 집계하기 (Level 3)

- **문제 출처**: [프로그래머스 카테고리 별 도서 판매량 집계하기](https://school.programmers.co.kr/learn/courses/30/lessons/144855)

---

### 문제 설명

2022년 1월의 카테고리 별 도서 판매량을 합산하고, 카테고리(`CATEGORY`), 총 판매량(`TOTAL_SALES`) 리스트를 출력하는 SQL문을 작성해주세요.
결과는 카테고리명을 기준으로 오름차순 정렬해야 합니다.

---

### 예시 테이블 및 기대 출력 결과

#### BOOK (도서 정보)
| BOOK_ID | CATEGORY | AUTHOR_ID | PRICE | PUBLISHED_DATE |
| --- | --- | --- | --- | --- |
| 1 | 인문 | 1 | 10000 | 2020-01-01 |
| 2 | 경제 | 2 | 9000 | 2021-02-05 |
| 3 | 생활 | 3 | 11000 | 2021-03-11 |

#### BOOK_SALES (도서 판매 정보)
| BOOK_ID | SALES_DATE | SALES |
| --- | --- | --- |
| 1 | 2022-01-10 | 2 |
| 1 | 2022-01-20 | 3 |
| 2 | 2022-01-15 | 4 |
| 3 | 2022-02-01 | 5 |

#### 기대하는 출력 결과 (RESULT)
| CATEGORY | TOTAL_SALES |
| --- | --- |
| 경제 | 4 |
| 인문 | 5 |


```python
-- 코드를 입력하세요
SELECT A.CATEGORY, B.SALES AS TOTAL_SALES
FROM BOOK A JOIN (SELECT BOOK_ID,SUM(SALES) AS SALES FROM BOOK_SALES WHERE SAELS_DATE >= '2022-01' AND SAELS_DATE <= '2022-02' GROUP BY BOOK_ID)
B ON  A.BOOK_ID = B.BOOK_ID
ORDER BY A.CATEGORY ASC;
```

## 문제 3: GROUP BY / HAVING - 대여 횟수가 많은 자동차들의 월별 대여 횟수 구하기 (Level 3)

- **문제 출처**: [프로그래머스 대여 횟수가 많은 자동차들의 월별 대여 횟수 구하기](https://school.programmers.co.kr/learn/courses/30/lessons/151139)

---

### 문제 설명

`CAR_RENTAL_COMPANY_RENTAL_HISTORY` 테이블에서 대여 시작일을 기준으로 2022년 8월부터 2022년 10월까지 총 대여 횟수가 5회 이상인 자동차들에 대해 해당 기간 동안의 월별 자동차 ID별 총 대여 횟수(컬럼명: `RENTALS`) 리스트를 출력하는 SQL문을 작성해주세요.

결과는 월을 기준으로 오름차순 정렬하고, 월이 같다면 자동차 ID를 기준으로 내림차순 정렬해주세요. 특정 월의 총 대여 횟수가 0인 경우는 결과에서 제외합니다.

---

### 예시 테이블 및 기대 출력 결과

#### CAR_RENTAL_COMPANY_RENTAL_HISTORY (자동차 대여 기록 정보)
| HISTORY_ID | CAR_ID | START_DATE | END_DATE |
| --- | --- | --- | --- |
| 1 | 1 | 2022-08-01 | 2022-08-03 |
| 2 | 1 | 2022-08-05 | 2022-08-09 |
| 3 | 1 | 2022-09-01 | 2022-09-03 |
| 4 | 1 | 2022-09-10 | 2022-09-13 |
| 5 | 1 | 2022-10-01 | 2022-10-05 |
| 6 | 2 | 2022-08-15 | 2022-08-17 |

#### 기대하는 출력 결과 (RESULT)
| MONTH | CAR_ID | RENTALS |
| --- | --- | --- |
| 8 | 1 | 2 |
| 9 | 1 | 2 |
| 10 | 1 | 1 |


```python
SELECT MONTH(START_DATE) AS MONTH, CAR_ID, COUNT(*) AS RENTALS
FROM CAR_RENTAL_COMPANY_RENTAL_HISTORY
WHERE START_DATE >= '2022-08-01' AND START_DATE < '2022-11-01'
  AND CAR_ID IN (
      SELECT CAR_ID
      FROM CAR_RENTAL_COMPANY_RENTAL_HISTORY
      WHERE START_DATE >= '2022-08-01' AND START_DATE < '2022-11-01'
      GROUP BY CAR_ID
      HAVING COUNT(*) >= 5
  )
GROUP BY MONTH(START_DATE), CAR_ID
ORDER BY MONTH ASC, CAR_ID DESC;
```
