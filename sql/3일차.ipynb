# 3일차 SQL 학습: NULL 처리, GROUP BY/HAVING & JOIN

## [문제 1]: [프로그래머스 Lv.2 - NULL 처리하기](https://school.programmers.co.kr/learn/courses/30/lessons/59410)

### 문제 설명
입동하여 보호소에 들어온 동물의 정보를 담은 `ANIMAL_INS` 테이블입니다.
동물 보호소에 들어온 동물의 생물 종(`ANIMAL_TYPE`), 이름(`NAME`), 성별 및 중성화 여부(`SEX_UPON_INTAKE`)를 아이디(`ANIMAL_ID`) 순으로 조회하는 SQL문을 작성해주세요.
이때 이름이 없는 채로 입소한 동물의 이름은 "No name"으로 표시해주세요.

---

#### 예시 테이블 데이터

##### ANIMAL_INS
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | INTAKE_CONDITION | NAME | SEX_UPON_INTAKE |
| :--- | :--- | :--- | :--- | :--- | :--- |
| A350276 | Cat | 2017-08-13 13:50:00 | Normal | Jewel | Spayed Female |
| A350375 | Cat | 2017-03-06 15:01:00 | Normal | NULL | Spayed Female |
| A368930 | Dog | 2014-06-08 13:20:00 | Normal | NULL | Spayed Female |
| A370507 | Cat | 2014-10-27 14:43:00 | Normal | Emily | Spayed Female |

##### 출력 예시 결과 (Expected Output)
| ANIMAL_TYPE | NAME | SEX_UPON_INTAKE |
| :--- | :--- | :--- |
| Cat | Jewel | Spayed Female |
| Cat | No name | Spayed Female |
| Dog | No name | Spayed Female |
| Cat | Emily | Spayed Female |

---

#### 과거 기록 기반 주의할 문제점 복습 노트
- **NULL 값 대체 (IFNULL / COALESCE / CASE WHEN)**:
  - MySQL에서 컬럼이 NULL인 경우 다른 값으로 치환할 때 `IFNULL(NAME, 'No name')` 또는 `COALESCE(NAME, 'No name')`을 사용합니다.
  - `CASE WHEN NAME IS NULL THEN 'No name' ELSE NAME END`로도 작성할 수 있습니다.

#### 힌트 & 생각 유도 가이드
- `ANIMAL_INS` 테이블에서 `ANIMAL_TYPE`, `IFNULL(NAME, 'No name') AS NAME`, `SEX_UPON_INTAKE` 컬럼을 `SELECT`합니다.
- 정렬 기준은 `ANIMAL_ID` 오름차순입니다.

```sql
-- 코드를 입력하세요
SELECT ANIMAL_TYPE, IFNULL(NAME, "No name") AS NAME, SEX_UPON_INTAKE
FROM ANIMAL_INS
ORDER BY ANIMAL_ID ASC;
```

## [문제 2]: [프로그래머스 Lv.2 - 동명 동물 수 찾기](https://school.programmers.co.kr/learn/courses/30/lessons/59041)

### 문제 설명
동물 보호소에 들어온 동물의 정보를 담은 `ANIMAL_INS` 테이블입니다.
동물 보호소에 들어온 동물 이름 중 두 번 이상 쓰인 이름과 해당 이름이 쓰인 횟수를 조회하는 SQL문을 작성해주세요.
이때 결과는 이름이 없는 동물(`NULL`)은 제외하며, 결과는 이름 순으로 정렬해주세요.

---

#### 예시 테이블 데이터

##### ANIMAL_INS
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | INTAKE_CONDITION | NAME | SEX_UPON_INTAKE |
| :--- | :--- | :--- | :--- | :--- | :--- |
| A399552 | Dog | 2013-10-14 15:38:00 | Normal | Jack | Neutered Male |
| A379968 | Dog | 2013-10-21 12:14:00 | Normal | Raven | Spayed Female |
| A354540 | Cat | 2014-12-11 11:48:00 | Normal | Lucy | Spayed Female |
| A353259 | Dog | 2014-04-10 13:22:00 | Normal | Lucy | Spayed Female |
| A354597 | Cat | 2014-05-02 12:12:00 | Normal | Lucy | Spayed Female |
| A379969 | Dog | 2014-05-02 12:12:00 | Normal | Raven | Spayed Female |

##### 출력 예시 결과 (Expected Output)
| NAME | COUNT |
| :--- | :--- |
| Lucy | 3 |
| Raven | 2 |

---

#### 과거 기록 기반 주의할 문제점 복습 노트
- **WHERE vs HAVING 구문 차이**:
  - `WHERE NAME IS NOT NULL`: 그룹화(`GROUP BY`) 전에 NULL 행을 먼저 필터링합니다.
  - `HAVING COUNT(NAME) >= 2` 또는 `HAVING COUNT(*) >= 2`: 그룹화 후 각 그룹의 행 개수가 2개 이상인 그룹만 집계 조건으로 필터링합니다.

#### 힌트 & 생각 유도 가이드
- `NAME`으로 그룹화(`GROUP BY NAME`)합니다.
- `WHERE NAME IS NOT NULL` 조건으로 이름이 없는 행을 사전에 제외합니다.
- `HAVING COUNT(NAME) >= 2`로 2번 이상 쓰인 이름만 걸러냅니다.
- `ORDER BY NAME ASC`로 이름 기준 오름차순 정렬합니다.

```sql
-- 코드를 입력하세요
SELECT NAME, COUNT(NAME) AS count
FROM ANIMAL_INS
WHERE NAME IS NOT NULL
GROUP BY NAME
HAVING COUNT(NAME) >=2
ORDER BY NAME ASC;
```

## [문제 3]: [프로그래머스 Lv.3 - 있었는데요 없었습니다](https://school.programmers.co.kr/learn/courses/30/lessons/59043)

### 문제 설명
`ANIMAL_INS` 테이블은 동물 보호소에 들어온 동물의 정보이고, `ANIMAL_OUTS` 테이블은 동물 보호소에서 입양 보낸 동물의 정보입니다.
관리자의 실수로 일부 동물의 입양일이 잘못 입력되었습니다. 보호 시작일(`ANIMAL_INS.DATETIME`)보다 입양일(`ANIMAL_OUTS.DATETIME`)이 더 빠른 동물의 아이디와 이름을 조회하는 SQL문을 작성해주세요.
이때 결과는 보호 시작일이 빠른 순(오름차순)으로 정렬해야 합니다.

---

#### 예시 테이블 데이터

##### ANIMAL_INS
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | INTAKE_CONDITION | NAME | SEX_UPON_INTAKE |
| :--- | :--- | :--- | :--- | :--- | :--- |
| A350276 | Cat | 2017-08-13 13:50:00 | Normal | Jewel | Spayed Female |
| A381217 | Dog | 2017-10-23 15:14:00 | Normal | Cherish | Neutered Male |
| A298750 | Cat | 2016-10-22 14:42:00 | Normal | NULL | Spayed Female |

##### ANIMAL_OUTS
| ANIMAL_ID | ANIMAL_TYPE | DATETIME | NAME | SEX_UPON_OUTCOME |
| :--- | :--- | :--- | :--- | :--- |
| A350276 | Cat | 2017-08-11 14:40:00 | Jewel | Spayed Female |
| A381217 | Dog | 2017-10-23 15:20:00 | Cherish | Neutered Male |
| A298750 | Cat | 2016-10-22 14:40:00 | NULL | Spayed Female |

##### 출력 예시 결과 (Expected Output)
| ANIMAL_ID | NAME |
| :--- | :--- |
| A350276 | Jewel |
| A298750 | NULL |

---

#### 과거 기록 기반 주의할 문제점 복습 노트
- **날짜/시간 비교 연산자 사용법**:
  - SQL에서 `DATETIME` 타입 데이터는 직접 비교 연산자(`<`, `>`)를 통해 이전/이후 비교가 가능합니다.
  - `ANIMAL_OUTS.DATETIME < ANIMAL_INS.DATETIME` 조건은 "입양일이 보호 시작일보다 더 빠른(이전인)" 경우를 의미합니다.

#### 힌트 & 생각 유도 가이드
- `ANIMAL_INS`와 `ANIMAL_OUTS` 테이블을 공통 키인 `ANIMAL_ID`로 `JOIN`합니다.
- `WHERE ANIMAL_OUTS.DATETIME < ANIMAL_INS.DATETIME` 조건으로 잘못 입력된 행을 필터링합니다.
- `SELECT ANIMAL_INS.ANIMAL_ID, ANIMAL_INS.NAME`을 선택합니다.
- `ORDER BY ANIMAL_INS.DATETIME ASC`로 보호 시작일 오름차순 정렬합니다.

```sql
-- 코드를 입력하세요
SELECT A.ANIMAL_ID, A.NAME
FROM ANIMAL_INS A JOIN ANIMAL_OUTS B ON A.ANIMAL_ID = B.ANIMAL_ID
WHERE B.DATETIME < A.DATETIME
ORDER BY A.DATETIME ASC;
```
