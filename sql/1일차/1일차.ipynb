## [문제 1]: [프로그래머스 Lv.2 - 자동차 평균 대여 기간 구하기](https://school.programmers.co.kr/learn/courses/30/lessons/157345)

### 문제 설명
`CAR_RENTAL_COMPANY_RENTAL_HISTORY` 테이블에서 평균 대여 기간이 **7일 이상인 자동차**들의 자동차 ID(`CAR_ID`)와 평균 대여 기간(`AVERAGE_DURATION`) 리스트를 출력하는 SQL문을 작성해주세요.

- 평균 대여 기간은 소수점 두 번째 자리에서 **반올림하여 첫째 자리까지** 표시하세요.
- 결과는 **평균 대여 기간을 기준으로 내림차순** 정렬하고, 평균 대여 기간이 같다면 **자동차 ID를 기준으로 내림차순** 정렬해주세요.

---

### 예시 테이블 데이터

#### CAR_RENTAL_COMPANY_RENTAL_HISTORY
| HISTORY_ID | CAR_ID | START_DATE | END_DATE |
| :--- | :--- | :--- | :--- |
| 1 | 1 | 2022-09-27 | 2022-10-01 |
| 2 | 2 | 2022-10-03 | 2022-10-04 |
| 3 | 2 | 2022-10-13 | 2022-10-15 |
| 4 | 1 | 2022-10-11 | 2022-10-16 |
| 5 | 3 | 2022-10-16 | 2022-10-17 |

#### 출력 예시 결과 (Expected Output)
| CAR_ID | AVERAGE_DURATION |
| :--- | :--- |
| 1 | 5.5 |

---

### 과거 기록 기반 주의할 문제점 복습 노트
1. **DATEDIFF와 시작일 포함 문제**: `DATEDIFF(END_DATE, START_DATE) + 1` 이용
2. **반올림(ROUND)과 평균(AVG) 순서**: `ROUND(AVG(...), 1)` 순서 준수
3. **WHERE vs HAVING 구별**: 그룹화 후 평균 조건은 `HAVING` 절 작성

### 힌트 & 생각 유도 가이드
- 대여 기간 = `DATEDIFF(END_DATE, START_DATE) + 1` 
- 평균 대여 기간 반올림 = `ROUND(AVG(...), 1)` 
- 자동차별 그룹화 = `GROUP BY CAR_ID` 
- 평균 대여 기간 7일 이상 필터링 = `HAVING AVG(...) >= 7` 
- 정렬 = `ORDER BY AVERAGE_DURATION DESC, CAR_ID DESC`

````python
-- 코드를 입력하세요
SELECT CAR_ID, ROUND(AVG(DATEDIFF(END_DATE, START_DATE)+1),1) AS AVERAGE_DURATION
FROM CAR_RENTAL_COMPANY_RENTAL_HISTORY
GROUP BY CAR_ID
HAVING AVG(DATEDIFF(END_DATE,START_DATE)+1) >= 7
ORDER BY AVERAGE_DURATION DESC, CAR_ID DESC;
````

## [문제 2]: [프로그래머스 Lv.2 - 조건에 부합하는 중고거래 상태 조회하기](https://school.programmers.co.kr/learn/courses/30/lessons/164672)

### 문제 설명
`USED_GOODS_BOARD` 테이블에서 **2022년 10월 5일에 등록된** 중고거래 게시글의 게시글 ID(`BOARD_ID`), 작성자 ID(`WRITER_ID`), 게시글 제목(`TITLE`), 가격(`PRICE`), 거래상태(`STATUS`)를 조회하는 SQL문을 작성해주세요.

- 거래상태가 `SALE` 이면 **'판매중'**, `RESERVED` 이면 **'예약중'**, `DONE` 이면 **'거래완료'** 로 출력하세요.
- 결과는 게시글 ID를 기준으로 내림차순 정렬해주세요.

---

### 예시 테이블 데이터

##### USED_GOODS_BOARD

### 과거 기록 기반 주의할 문제점 복습 노트
1. **CASE WHEN 문법 지키기**: 구문의 맨 끝에는 반드시 `END` 키워드를 붙여야 하며 컴마(`,`) 금지
2. **날짜 조건 검사**: `DATE_FORMAT(CREATED_DATE, '%Y-%m-%d') = '2022-10-05'` 처럼 비교

### 힌트 & 생각 유도 가이드
- CASE 구문 형태:
  ```sql
  CASE STATUS 
      WHEN 'SALE' THEN '판매중'
      WHEN 'RESERVED' THEN '예약중'
      WHEN 'DONE' THEN '거래완료'
  END AS STATUS
  ```
- 날짜 조건 = `WHERE CREATED_DATE = '2022-10-05'` 또는 `DATE_FORMAT(...)` 사용
- 정렬 = `ORDER BY BOARD_ID DESC`

````python
-- 코드를 입력하세요
SELECT BOARD_ID,WRITER_ID,TITLE,PRICE,
case STATUS
when 'SALE' THEN '판매중'
WHEN 'RESERVED' then '예약중'
WHEN 'DONE' THEN '거래완료' 
END AS STATUS
From used_goods_board
where CREATED_DATE = '2022-10-05'
order by board_id desc;
````
